DESAFIO 19: ORDENAR EL INVENTARIO DE LIBROS:
Para ordenar los libros por sus códigos de identificación en orden decreciente sin modificar la lista original, debes usar la función sorted() con el parámetro reverse=True:

```python
codigos_libros = [104, 302, 101, 505, 208]

# Orden decreciente creando una nueva lista
libros_ordenados = sorted(codigos_libros, reverse=True)

print("Original:", codigos_libros)      # [104, 302, 101, 505, 208]
print("Ordenados:", libros_ordenados)    # [505, 302, 208, 104, 101]
```

¿Por qué es importante no modificar la lista original?
Preservación del registro histórico: La lista original puede representar el orden cronológico real de llegada de los libros a la biblioteca o la ubicación física original en los estantes. Al alterarla, perderías esa información de origen.

Efectos secundarios no deseados (Side Effects): En programas más grandes, otras funciones o partes del sistema pueden depender del orden original de los datos. Modificar la estructura en un sitio puede generar errores en otras secciones del código.

Inmutabilidad y seguridad: Trabajar con una copia ordenada te permite realizar consultas temporales sin arriesgar la integridad de los datos maestros.
¿Por qué no puedo usar el método .sort() sobre la lista original?
Modificación in situ (In-place): El método .sort() reordena los elementos directamente dentro de la lista original (la muta), destruyendo la secuencia inicial.

Retorna None: A diferencia de sorted(), el método .sort() no devuelve una nueva lista, sino que devuelve None. Si intentas hacer copia = lista.sort(), la variable copia almacenará None y la lista original ya habrá sido alterada.

DESAFIO 20: caracteres ASCII: Para resolver este desafío y convertir una lista de códigos numéricos ASCII a sus respectivos caracteres, utilizamos la función integrada chr() dentro de un bucle o una list de comprensión

```python
# Lista de códigos ASCII de prueba
codigos_ascii = [72, 111, 108, 97, 33]

# Opción 1: Usando comprensión de listas (la forma más directa en Python)
caracteres = [chr(codigo) for codigo in codigos_ascii]

# Opción 2: Uniendo los caracteres para formar un texto visible
texto_resultado = "".join(caracteres)

print("Lista de caracteres:", caracteres)
print("Texto resultante:", texto_resultado)
```

DENOMINACIONES IMPORTANTES:
chr(numero): Recibe un número entero que representa el código Unicode/ASCII y devuelve una cadena de texto (string) con el carácter correspondiente (por ejemplo, chr(65) devuelve 'A').

ord(caracter): Es la función inversa a chr(). Recibe un carácter de texto y devuelve su código entero ASCII/Unicode (por ejemplo, ord('A') devuelve 65).

.join(): Permite concatenar los elementos de una lista de caracteres en una sola string continua para leer la palabra completa.

DESAFIO 21: CONTROL DE NOTAS DE ESTUDIANTES:
Pregunta 1: Calcular el promedio
Se suma el total de las notas con sum() y se divide entre la cantidad total de estudiantes con len():

```python
notas = [7.5, 9.0, 6.0, 4.5, 9.0, 8.0, 6.0, 9.0, 5.5, 7.0]

promedio = sum(notas) / len(notas)
print(f"Promedio de la clase: {promedio:.2f}")  # Resultado: 7.15
```

Pregunta 2: Nota más baja y más alta
Utilizas las funciones integradas min() y max():

```python
nota_minima = min(notas)
nota_maxima = max(notas)

print(f"Nota más baja: {nota_minima}")  # Resultado: 4.5
print(f"Nota más alta: {nota_maxima}")  # Resultado: 9.0
```

Pregunta 3: Nota que más se repite (Moda)
Usas Counter de la librería collections y su método .most_common(1) para obtener la nota con mayor frecuencia:

```python
from collections import Counter

conteo = Counter(notas)
# most_common(1) devuelve una lista con la tupla (elemento, frecuencia)
nota_mas_repetida, frecuencia = conteo.most_common(1)[0]

print(f"La nota que más se repite es {nota_mas_repetida} (se repite {frecuencia} veces)")
```

Pregunta 4 (Plus): Gráfico de barras de frecuencias
Con matplotlib.pyplot, agrupamos la frecuencia de cada nota y generamos el gráfico de barras:

```python
import matplotlib.pyplot as plt
from collections import Counter

# 1. Obtener la frecuencia de cada nota
conteo_notas = Counter(notas)
valores_notas = list(conteo_notas.keys())
frecuencias = list(conteo_notas.values())

# 2. Crear el gráfico de barras
plt.bar(valores_notas, frecuencias, width=0.4, color='skyblue', edgecolor='black')

# 3. Personalizar el gráfico
plt.title("Distribución de Notas de la Clase")
plt.xlabel("Notas")
plt.ylabel("Cantidad de Estudiantes")
plt.grid(axis='y', linestyle='--', alpha=0.7)

# 4. Mostrar el gráfico
plt.show()
```
