Desafío 81: Recorrido de estudiantes por niveles

Dado un árbol que representa los grupos de estudiantes en una escuela, implementa un recorrido por niveles para mostrar los estudiantes de cada grupo, comenzando por el nivel más alto (ej. grado 12) y descendiendo hasta el nivel más bajo (ej. grado 1). Cada nodo del árbol representa un grado y sus estudiantes.



```python
from collections import deque

class NodoGrado:
    def __init__(self, grado, estudiantes):
        self.grado = grado
        self.estudiantes = estudiantes  # Lista de nombres de estudiantes
        self.hijos = []                 # Lista de sub-grados (nodos hijos)

def recorrido_por_niveles(raiz):
    if not raiz:
        return

    cola = deque([raiz])
    nivel = 1

    while cola:
        n = len(cola)
        print(f"--- Nivel {nivel} ---")

        for _ in range(n):
            nodo = cola.popleft()
            print(f"Grado: {nodo.grado} | Estudiantes: {', '.join(nodo.estudiantes)}")

            # Agregamos los sub-grados (hijos) a la cola para el siguiente nivel
            for hijo in nodo.hijos:
                cola.append(hijo)

        nivel += 1

# Ejemplo de uso:
# raiz = NodoGrado("Grado 12", ["Ana", "Carlos"])
# recorrido_por_niveles(raiz)
```

Recorrido de estudiantes por niveles
Para recorrer un árbol por niveles (también conocido como búsqueda en anchura o BFS), podemos utilizar una cola (deque) para visitar los nodos nivel por nivel, desde la raíz hasta las hojas.

Desafío 82: Implementar búsqueda secuencial en una tabla de calificaciones

Tienes una tabla de calificaciones representada como una matriz, donde cada fila contiene las calificaciones de un estudiante en distintas materias. Implementa una función que busque una calificación específica en toda la matriz y devuelva el estudiante y la materia en la que se encuentra.

```python
def buscar_calificacion(matriz_calificaciones, lista_estudiantes, lista_materias, objetivo):
    """
    Busca una calificación específica en la matriz.
    matriz_calificaciones: lista de listas (filas = estudiantes, columnas = materias)
    """
    for i, fila in enumerate(matriz_calificaciones):
        for j, calificacion in enumerate(fila):
            if calificacion == objetivo:
                estudiante = lista_estudiantes[i]
                materia = lista_materias[j]
                return f"Calificación {objetivo} encontrada: Estudiante '{estudiante}', Materia '{materia}' (Fila {i}, Columna {j})."

    return f"La calificación {objetivo} no se encuentra en la tabla."

# Ejemplo de uso:
estudiantes = ["Sofía", "Mateo", "Lucía"]
materias = ["Matemáticas", "Historia", "Programación"]
matriz = [
    [85, 90, 78],
    [92, 88, 95],
    [70, 75, 80]
]

# print(buscar_calificacion(matriz, estudiantes, materias, 95))
```

Desafío 83: Optimizar la búsqueda en una lista ordenada de estudiantes

Tienes una lista ordenada alfabéticamente con los nombres de los estudiantes de una clase. Implementa una función que realice una búsqueda binaria para encontrar un estudiante específico en la lista. Si el estudiante no está, la función debe mostrar un mensaje adecuado.

```python
def busqueda_binaria_estudiante(lista_ordenada, estudiante_buscado):
    izquierda = 0
    derecha = len(lista_ordenada) - 1

    while izquierda <= derecha:
        medio = (izquierda + derecha) // 2

        if lista_ordenada[medio] == estudiante_buscado:
            return f"¡Estudiante encontrado! '{estudiante_buscado}' está en la posición {medio}."
        elif lista_ordenada[medio] < estudiante_buscado:
            izquierda = medio + 1
        else:
            derecha = medio - 1

    return f"El estudiante '{estudiante_buscado}' no se encuentra en la lista."

# Ejemplo de uso:
estudiantes_ordenados = ["Ana", "Bruno", "Carla", "Daniel", "Elena"]
# print(busqueda_binaria_estudiante(estudiantes_ordenados, "Carla"))
# print(busqueda_binaria_estudiante(estudiantes_ordenados, "Marcos"))
```

Optimizar la búsqueda en una lista ordenada de estudiantes:
La búsqueda binaria divide repetidamente el espacio de búsqueda a la mitad, lo que requiere que la lista esté previamente ordenada alfabéticamente.

Desafío 84: Ordenar estudiantes por promedio de calificaciones

Tienes una lista de estudiantes y su promedio de calificaciones. Implementa un algoritmo que ordene a los estudiantes de acuerdo con su promedio utilizando el algoritmo de ordenamiento por selección. Al final, el estudiante con el promedio más alto debe estar en primer lugar.

```python
def ordenar_estudiantes_por_promedio(lista_estudiantes):
    """
    lista_estudiantes: lista de tuplas con formato [('Nombre', promedio), ...]
    """
    n = len(lista_estudiantes)

    for i in range(n):
        max_idx = i
        for j in range(i + 1, n):
            # Buscamos el mayor promedio para colocarlo al inicio
            if lista_estudiantes[j][1] > lista_estudiantes[max_idx][1]:
                max_idx = j

        # Intercambiamos el elemento actual con el mayor encontrado
        lista_estudiantes[i], lista_estudiantes[max_idx] = lista_estudiantes[max_idx], lista_estudiantes[i]

    return lista_estudiantes

# Ejemplo de uso:
alumnos = [("Juan", 7.5), ("María", 9.2), ("Pedro", 8.1), ("Lucía", 9.8)]
# alumnos_ordenados = ordenar_estudiantes_por_promedio(alumnos)
# print(alumnos_ordenados)
```

Ordenar estudiantes por promedio de calificaciones:
Se implementa el algoritmo de ordenamiento por selección (Selection Sort) adaptado para ordenar los promedios de forma descendente (del mayor al menor promedio).

Dado un conjunto de estudiantes y sus promedios, implementa una función que cree un árbol binario de búsqueda en el que los nodos representan los promedios de los estudiantes. Luego, implementa una función que recorra el árbol en inorden para mostrar los estudiantes en orden ascendente de rendimiento académico.

```python
class NodoBST:
    def __init__(self, promedio, estudiante):
        self.promedio = promedio
        self.estudiantes = [estudiante]  # Lista por si hay promedios duplicados
        self.izq = None
        self.der = None

def insertar_en_bst(raiz, promedio, estudiante):
    if not raiz:
        return NodoBST(promedio, estudiante)

    if promedio < raiz.promedio:
        raiz.izq = insertar_en_bst(raiz.izq, promedio, estudiante)
    elif promedio > raiz.promedio:
        raiz.der = insertar_en_bst(raiz.der, promedio, estudiante)
    else:
        # Si el promedio es exactamente igual, agrupamos al estudiante en el mismo nodo
        raiz.estudiantes.append(estudiante)

    return raiz

def recorrido_inorden(raiz, resultado=None):
    if resultado is None:
        resultado = []

    if raiz:
        recorrido_inorden(raiz.izq, resultado)
        for est in raiz.estudiantes:
            resultado.append((est, raiz.promedio))
        recorrido_inorden(raiz.der, resultado)

    return resultado

# Ejemplo de uso:
# raiz = None
# datos = [("Lucas", 8.5), ("Valeria", 7.0), ("Mateo", 9.1), ("Camila", 8.5)]
# for est, prom in datos:
#     raiz = insertar_en_bst(raiz, prom, est)

# rendimiento_ordenado = recorrido_inorden(raiz)
# print("Rendimiento ascendente:", rendimiento_ordenado)
```

Crear un árbol de clasificación de estudiantes por rendimiento:
