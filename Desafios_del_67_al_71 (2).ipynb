Desafio 68:Crea un programa que tome una lista de valores y realice operaciones matemáticas sobre ellos. Implementa el manejo de varias excepciones comunes como ZeroDivisionError, TypeError, y ValueError.

```python
def procesar_operaciones(valores):
    print(f"Lista a procesar: {valores}\n")

    for item in valores:
        try:
            print(f"Procesando elemento: {item}")

            # Intentamos convertir el elemento a entero.
            # Esto puede disparar ValueError (si es texto no numérico)
            # o TypeError (si es un tipo no compatible como None).
            numero = int(item)

            # Realizamos una operación matemática (división)
            # Esto puede disparar ZeroDivisionError si el número es 0.
            resultado = 100 / numero
            print(f"-> Éxito: 100 / {numero} = {resultado}\n")

        except ValueError:
            print(f"❌ Error de valor (ValueError): '{item}' no se pudo convertir a entero.\n")
        except ZeroDivisionError:
            print("❌ Error matemático (ZeroDivisionError): No se puede dividir entre cero.\n")
        except TypeError:
            print(f"❌ Error de tipo (TypeError): El tipo de dato '{type(item).__name__}' no es compatible.\n")

# Lista de prueba con elementos variados para activar cada excepción
mi_lista = [20, "texto_falso", 0, "50", None]

procesar_operaciones(mi_lista)
```

De esta forma interactúan las excepciones:
ValueError: Se dispara cuando int("texto_falso") intenta convertir un texto que no tiene formato numérico.

ZeroDivisionError: Se dispara cuando el valor procesado es 0 y se intenta hacer 100 / 0.

TypeError: Se dispara cuando intentamos operar con un tipo de dato no admitido por la función de conversión (por ejemplo, int(None) genera un TypeError en Python).

Desafio 69: Escribe una función que calcule el factorial de un número entero positivo. Maneja las excepciones si el número ingresado es negativo, no es entero, o es demasiado grande para ser procesado.

```python
import math

def calcular_factorial(entrada):
    try:
        # 1. Validar si es un valor convertible a número y verificar si es entero
        # Verificamos si la entrada contiene un punto decimal (por ejemplo, "5.4" o float)
        if isinstance(entrada, float) or (isinstance(entrada, str) and "." in entrada):
            # Intentamos convertir a float primero para ver si tiene decimales reales
            num_float = float(entrada)
            if not num_float.is_integer():
                raise ValueError("El número no es un entero.")
            numero = int(num_float)
        else:
            numero = int(entrada) # Si es entero o string numérico entero

        # 2. Manejar si el número es negativo
        if numero < 0:
            raise ValueError("El número no puede ser negativo.")

        # 3. Calcular el factorial utilizando math.factorial
        # math.factorial lanza un ValueError automático si el número es negativo,
        # pero ya lo manejamos arriba. Para números "demasiado grandes",
        # Python maneja enteros arbitrarios, pero si excede el límite de recursión
        # o memoria/tiempo del sistema, puede lanzar un OverflowError o MemoryError.
        # También podemos poner un límite personalizado si consideramos "demasiado grande" (ej. > 2000).
        if numero > 2000:
            raise OverflowError("El número es demasiado grande para ser procesado de forma eficiente.")

        resultado = math.factorial(numero)
        return f"El factorial de {numero} es: {resultado}"

    except ValueError as e:
        # Atrapa tanto errores de formato (letras) como nuestras validaciones de enteros y negativos
        return f"❌ Error de valor: {e}"
    except OverflowError as e:
        # Atrapa si el número excede el límite razonable de procesamiento
        return f"❌ Error de desbordamiento: {e}"
    except TypeError:
        # Atrapa si pasan un tipo de dato completamente incompatible (como una lista o None)
        return "❌ Error de tipo: El valor ingresado no es válido."

# --- Pruebas del programa ---
print(calcular_factorial(5))       # Caso exitoso: 120
print(calcular_factorial(-3))      # Caso negativo
print(calcular_factorial("4.5"))   # Caso no entero
print(calcular_factorial("abc"))   # Caso no numérico
print(calcular_factorial(3000))    # Caso demasiado grande
```

Explicación de los manejos de excepciones:
No es entero o no es numérico (ValueError / TypeError): Verificamos si la entrada tiene decimales o caracteres extraños mediante conversiones y validamos con bloques except.

Número negativo (ValueError): Se lanza una excepción controlada si el número es menor que 0, ya que el factorial no está definido para negativos (en los reales).

Demasiado grande (OverflowError): Aunque Python soporta enteros de tamaño arbitrario en memoria, calcular factoriales de números gigantescos (como 100,000) puede congelar el programa o agotar los recursos. Establecemos un límite preventivo (por ejemplo, > 2000) para lanzar un OverflowError.

Desafio 70: Desafío 70:

Crea una excepción personalizada llamada NegativeNumberError que se dispare si el usuario intenta ingresar un número negativo en un programa de cálculo de raíces cuadradas. Implementa el manejo de esta excepción en el programa.

```python
import math

# 1. Definimos la excepción personalizada
class NegativeNumberError(Exception):
    """Excepción lanzada cuando se intenta ingresar un número negativo."""
    pass

def calcular_raiz_cuadrada():
    try:
        # Solicitamos el número al usuario
        entrada = input("Ingresa un número para calcular su raíz cuadrada: ")

        # Convertimos a float (o int) para permitir decimales
        numero = float(entrada)

        # 2. Verificamos si es negativo para lanzar nuestra excepción personalizada
        if numero < 0:
            raise NegativeNumberError("No se puede calcular la raíz cuadrada real de un número negativo.")

        # Calculamos la raíz cuadrada si todo es correcto
        resultado = math.sqrt(numero)
        print(f"La raíz cuadrada de {numero} es: {resultado}")

    except ValueError:
        # Captura si el usuario ingresa letras u otros caracteres no numéricos
        print("Error: Debes ingresar un valor numérico válido.")

    except NegativeNumberError as e:
        # Captura nuestra excepción personalizada y muestra su mensaje
        print(f"Error personalizado: {e}")

# Ejecutamos la función
calcular_raiz_cuadrada()
```

Al realizar esta actividad es importante tener en cuenta:
Creación de la excepción: class NegativeNumberError(Exception): pass crea una nueva excepción heredando de la clase general Exception de Python.

Uso de raise: La palabra clave raise nos permite interrumpir el flujo normal del programa y disparar nuestra excepción de manera controlada cuando detectamos la condición no deseada (numero < 0).

Captura específica: El bloque except NegativeNumberError as e: atrapa exclusivamente nuestro error personalizado y nos permite usar el mensaje que definimos.

Desafio 71: Desarrolla un programa que abra un archivo de texto, lea su contenido y lo muestre. Implementa manejo de excepciones para archivos que no existan y usa finally para asegurarte de que el archivo se cierre adecuadamente en cualquier caso.

```python
def leer_archivo_texto(nombre_archivo):
    archivo = None  # Inicializamos la variable para el control en el finally
    try:
        # Intentamos abrir el archivo en modo lectura ("r") con codificación utf-8
        archivo = open(nombre_archivo, "r", encoding="utf-8")

        # Leemos y mostramos el contenido
        contenido = archivo.read()
        print(f"--- Contenido de '{nombre_archivo}' ---")
        print(contenido)

    except FileNotFoundError:
        # Capturamos el error si el archivo no existe
        print(f"❌ Error: El archivo '{nombre_archivo}' no fue encontrado.")

    except PermissionError:
        # Capturamos el error si no hay permisos de lectura
        print(f"❌ Error: No tienes permisos para leer el archivo '{nombre_archivo}'.")

    finally:
        # El bloque finally SE EJECUTA SIEMPRE, ocurra o no una excepción
        if archivo is not None and not archivo.closed:
            archivo.close()
            print("🔒 El archivo ha sido cerrado de forma segura en el bloque finally.")
        else:
            print("🔒 El archivo nunca llegó a abrirse, por lo que no requirió cierre.")

# --- Ejemplo de prueba ---
# Probamos con un archivo que no existe para ver el manejo de excepciones y el finally
leer_archivo_texto("documento_inexistente.txt")
```

Tener en cuenta para este desafío:
open(..., "r"): Abre el archivo en modo de lectura. Si el archivo no existe, Python genera automáticamente un FileNotFoundError.

Bloque finally: Garantiza que las acciones de limpieza (como cerrar conexiones a bases de datos o archivos abiertos) se ejecuten sin importar si el código funcionó correctamente o si falló y saltó al bloque except.

Validación archivo.closed: Nos aseguramos de que la variable exista y esté abierta antes de intentar cerrarla para evitar un error secundario de tipo NameError o AttributeError.
