Desafío 86: Crear un archivo de préstamos de libros

Desarrolla un programa que cree un archivo prestamos.txt y permita al usuario agregar el registro de un préstamo. El registro debe incluir el nombre del libro, el nombre del prestatario y la fecha del préstamo. Asegúrate de no sobrescribir el archivo cada vez que se agrega un nuevo préstamo.

```python
def registrar_prestamo(nombre_libro, prestatario, fecha):
    """
    Agrega un nuevo registro de préstamo al archivo 'prestamos.txt'
    sin sobrescribir los datos existentes.
    """
    with open("prestamos.txt", "a", encoding="utf-8") as archivo:
        archivo.write(f"Libro: {nombre_libro} | Prestatario: {prestatario} | Fecha: {fecha}\n")

    print("¡Préstamo registrado exitosamente!")

# Ejemplo de uso:
# registrar_prestamo("Cien años de soledad", "María Pérez", "2026-10-27")
```

Crear un archivo de préstamos de libros.
Para agregar registros sin sobrescribir el contenido previo, abrir el archivo en modo apéndice ("a" de append).

Desafío 87: Buscar un libro por autor en un archivo de texto

Dado un archivo libros.txt que contiene una lista de libros y sus autores, implementa una función que busque todos los libros escritos por un autor específico y los muestre. Si el autor no tiene libros en la lista, debe mostrar un mensaje indicando que no hay coincidencias.

```python
def buscar_libros_por_autor(nombre_archivo, autor_buscado):
    encontrados = False
    try:
        with open(nombre_archivo, "r", encoding="utf-8") as archivo:
            print(f"Buscando libros del autor: '{autor_buscado}'...")
            for linea in archivo:
                if "-" in linea:
                    libro, autor = linea.strip().split("-", 1)
                    if autor.strip().lower() == autor_buscado.strip().lower():
                        print(f"-> {libro.strip()}")
                        encontrados = True

        if not encontrados:
            print("No se encontraron coincidencias para este autor en la lista.")
    except FileNotFoundError:
        print(f"El archivo '{nombre_archivo}' no fue encontrado.")

# Ejemplo de uso:
# buscar_libros_por_autor("libros.txt", "Gabriel García Márquez")
```

Buscar un libro por autor en un archivo de texto.
Este programa lee línea por línea un archivo (asumiendo el formato Título - Autor) y filtra las coincidencias ignorando mayúsculas/minúsculas.

Desafío 88: Actualizar la cantidad de libros disponibles en un archivo CSV

Tienes un archivo inventario.csv que contiene una lista de libros y el número de copias disponibles. Escribe un programa que permita actualizar la cantidad de copias de un libro específico. El programa debe leer el archivo, modificar el número de copias y volver a escribir el archivo.

```python
import csv

def actualizar_inventario(nombre_archivo, titulo_libro, nueva_cantidad):
    filas = []
    encontrado = False

    try:
        # 1. Leer el archivo y buscar el libro
        with open(nombre_archivo, mode="r", encoding="utf-8") as archivo:
            lector = csv.reader(archivo)
            for fila in lector:
                if fila and fila[0].strip().lower() == titulo_libro.strip().lower():
                    fila[1] = str(nueva_cantidad)  # Actualiza la columna de copias
                    encontrado = True
                filas.append(fila)

        # 2. Si se encontró, reescribir el archivo con la nueva cantidad
        if encontrado:
            with open(nombre_archivo, mode="w", newline="", encoding="utf-8") as archivo:
                escritor = csv.writer(archivo)
                escritor.writerows(filas)
            print(f"Inventario actualizado. '{titulo_libro}' ahora tiene {nueva_cantidad} copias.")
        else:
            print(f"El libro '{titulo_libro}' no se encuentra en el inventario.")

    except FileNotFoundError:
        print(f"El archivo '{nombre_archivo}' no fue encontrado.")

# Ejemplo de uso (asumiendo formato: Título,Copias):
# actualizar_inventario("inventario.csv", "Python Básico", 5)
```

Actualizar la cantidad de libros disponibles en un archivo CSV.
Utiliza el módulo nativo csv para leer todas las filas, modifica la cantidad del libro seleccionado y reescribe el archivo completo.

Desafío 89: Eliminar registros de un archivo de préstamos

Escribe un programa que permita eliminar el registro de un préstamo de un archivo prestamos.txt. El programa debe mostrar los registros actuales, permitir al usuario seleccionar cuál eliminar, y luego actualizar el archivo sin el registro eliminado.



```python
def eliminar_prestamo(nombre_archivo):
    try:
        with open(nombre_archivo, "r", encoding="utf-8") as archivo:
            lineas = archivo.readlines()

        if not lineas:
            print("No hay registros de préstamos en el archivo.")
            return

        print("--- Registros Actuales de Préstamos ---")
        for i, linea in enumerate(lineas):
            print(f"[{i + 1}] {linea.strip()}")

        seleccion = int(input("Ingrese el número del registro que desea eliminar: ")) - 1

        if 0 <= seleccion < len(lineas):
            eliminado = lineas.pop(seleccion)
            with open(nombre_archivo, "w", encoding="utf-8") as archivo:
                archivo.writelines(lineas)
            print(f"Registro eliminado con éxito:\n-> {eliminado.strip()}")
        else:
            print("Número de registro inválido.")

    except FileNotFoundError:
        print(f"El archivo '{nombre_archivo}' no fue encontrado.")
    except ValueError:
        print("Error: Por favor, ingrese un número entero válido.")

# Ejemplo de uso:
# eliminar_prestamo("prestamos.txt")
```

Eliminar registros de un archivo de préstamos.
Mostra los registros actuales numerados, permite al usuario seleccionar uno mediante su índice y actualizar el archivo excluyendo dicho registro.

Desafío 90: Contar las palabras más comunes en un archivo de texto

Desarrolla una función que lea un archivo de texto grande, como libros.txt, y cuente cuántas veces aparece cada palabra. Luego, muestra las 5 palabras más comunes y cuántas veces aparecen.

```python
from collections import Counter
import re

def palabras_mas_comunes(nombre_archivo):
    try:
        with open(nombre_archivo, "r", encoding="utf-8") as archivo:
            texto = archivo.read().lower()

        # Extraemos solo palabras, omitiendo signos de puntuación y espacios
        palabras = re.findall(r'\b[a-záéíóúñ]+\b', texto)

        contador = Counter(palabras)
        top_5 = contador.most_common(5)

        print("--- Las 5 palabras más comunes ---")
        for posicion, (palabra, frecuencia) in enumerate(top_5, 1):
            print(f"{posicion}. Palabra: '{palabra}' -> Aparece {frecuencia} veces")

    except FileNotFoundError:
        print(f"El archivo '{nombre_archivo}' no fue encontrado.")

# Ejemplo de uso:
# palabras_mas_comunes("libros.txt")
```

Contar las palabras más comunes en un archivo de texto
Utiliza expresiones regulares (re) para extraer exclusivamente palabras limpias y collections.Counter para contabilizar eficientemente las 5 palabras más repetidas.
