Desafio N°52: Crea una clase Libro que tenga atributos privados para el título, autor y ISBN. Proporciona métodos getter y setter para cada atributo.

El principio de encapsulamiento no solo permite proteger los datos de un objeto frente a modificaciones indebidas, sino que asegura que la información mantenga su integridad mediante validaciones dentro de los setters.

Aunque el patrón tradicional basado en métodos get_ y set_ es estándar en lenguajes como Java o C++, en Python la forma idiomática recomendada es el uso de propiedades (@property), ya que ofrece la misma protección y control manteniendo una sintaxis mucho más limpia y natural.

DESAFIO N°54: Implementa la clase Autor con métodos getter y setter utilizando decoradores @property para manejar los atributos privados nombre y nacionalidad.

```python
class Libro:
    def __init__(self, titulo: str, isbn: str):
        self._titulo = titulo
        self._isbn = isbn

    @property
    def titulo(self) -> str:
        return self._titulo

    @property
    def isbn(self) -> str:
        return self._isbn

    def __str__(self) -> str:
        return f"'{self._titulo}' (ISBN: {self._isbn})"


class Autor:
    def __init__(self, nombre: str, nacionalidad: str):
        # Atributos privados (convención de Python con prefijo _)
        self._nombre = nombre
        self._nacionalidad = nacionalidad
        self._libros = []

    # --- Property para NOMBRE ---
    @property
    def nombre(self) -> str:
        """Getter para el nombre del autor."""
        return self._nombre

    @nombre.setter
    def nombre(self, valor: str) -> None:
        """Setter para el nombre del autor con validación."""
        if valor and valor.strip():
            self._nombre = valor.strip()
        else:
            raise ValueError("El nombre no puede estar vacío.")

    # --- Property para NACIONALIDAD ---
    @property
    def nacionalidad(self) -> str:
        """Getter para la nacionalidad del autor."""
        return self._nacionalidad

    @nacionalidad.setter
    def nacionalidad(self, valor: str) -> None:
        """Setter para la nacionalidad del autor con validación."""
        if valor and valor.strip():
            self._nacionalidad = valor.strip()
        else:
            raise ValueError("La nacionalidad no puede estar vacía.")

    # --- Property de solo lectura para la lista de LIBROS ---
    @property
    def libros(self) -> list:
        """Getter que retorna una copia de la lista para proteger el estado interno."""
        return self._libros.copy()

    # --- Método para agregar libros ---
    def agregar_libro(self, libro: Libro) -> None:
        if isinstance(libro, Libro):
            self._libros.append(libro)
        else:
            raise TypeError("El objeto a agregar debe ser una instancia de la clase Libro.")

    def mostrar_bibliografia(self) -> None:
        print(f"\nBibliografía de {self.nombre} ({self.nacionalidad}):")
        if not self._libros:
            print("  - Sin libros registrados.")
        else:
            for libro in self._libros:
                print(f"  - {libro}")


# Ejemplo de uso:
if __name__ == "__main__":
    # Instanciación
    autor = Autor("Julio Cortázar", "Argentina")

    # Acceso a atributos mediante @property (sintaxis natural de atributo)
    print(f"Autor: {autor.nombre}")
    print(f"Nacionalidad: {autor.nacionalidad}")

    # Modificación mediante setters (@nombre.setter y @nacionalidad.setter)
    autor.nombre = "Julio Florencio Cortázar"
    autor.nacionalidad = "Franco-Argentina"

    # Agregar libros
    autor.agregar_libro(Libro("Rayuela", "978-8437604572"))
    autor.agregar_libro(Libro("Bestiario", "978-8437602059"))

    # Resultado final
    autor.mostrar_bibliografia()
```

```python
class Autor:
    def __init__(self, nombre: str, nacionalidad: str):
        # Asignamos a través de las propiedades para aplicar las validaciones desde el inicio
        self.nombre = nombre
        self.nacionalidad = nacionalidad

    # --- Propiedad NOMBRE ---
    @property
    def nombre(self) -> str:
        return self._nombre

    @nombre.setter
    def nombre(self, valor: str) -> None:
        # Validación: comprueba que no sea None y que no sea una cadena vacía ni contenga solo espacios
        if isinstance(valor, str) and valor.strip():
            self._nombre = valor.strip()
        else:
            raise ValueError("El nombre no puede estar vacío.")

    # --- Propiedad NACIONALIDAD ---
    @property
    def nacionalidad(self) -> str:
        return self._nacionalidad

    @nacionalidad.setter
    def nacionalidad(self, valor: str) -> None:
        # Validación: comprueba que no sea None y que no sea una cadena vacía ni contenga solo espacios
        if isinstance(valor, str) and valor.strip():
            self._nacionalidad = valor.strip()
        else:
            raise ValueError("La nacionalidad no puede estar vacía.")


# ==========================================
# Demostración y Pruebas
# ==========================================
if __name__ == "__main__":
    # 1. Crear un objeto Autor
    autor1 = Autor("Jorge Luis Borges", "Argentina")

    # 2. Mostrar sus datos utilizando las propiedades (@property)
    print("--- Datos Iniciales ---")
    print(f"Nombre: {autor1.nombre}")
    print(f"Nacionalidad: {autor1.nacionalidad}\n")

    # 3. Modificar su nombre y nacionalidad mediante los setter
    autor1.nombre = "J. L. Borges"
    autor1.nacionalidad = "Argentino"

    print("--- Datos Modificados ---")
    print(f"Nuevo Nombre: {autor1.nombre}")
    print(f"Nueva Nacionalidad: {autor1.nacionalidad}\n")

    # 4. Intentar asignar valores vacíos
    print("--- Pruebas de Validación ---")

    try:
        autor1.nombre = ""
    except ValueError as error:
        print(f"Error capturado al asignar nombre vacío: {error}")

    try:
        autor1.nacionalidad = "   "  # Cadena con solo espacios en blanco
    except ValueError as error:
        print(f"Error capturado al asignar nacionalidad vacía: {error}")
```

¿Qué ventaja tiene usar @property frente a permitir el acceso directo a los atributos?

Control e Integridad de Datos (Validación): Permite validar los datos antes de guardarlos (por ejemplo, impidiendo cadenas vacías, valores None o formatos incorrectos) sin cambiar la forma en que los usuarios de la clase leen o asignan los atributos.

Encapsulamiento sin romper el código existente: Si al principio usas un atributo público como autor.nombre y más adelante necesitas agregar una lógica especial (como limpiar espacios con .strip() o registrar cambios), el uso de @property te permite añadir esa lógica sin obligar a modificar el código externo que ya usa la sintaxis autor.nombre.

Sintaxis limpia y "pythonica": Combina la seguridad de los métodos getter y setter tradicionales (como get_nombre() y set_nombre()) con una sintaxis limpia y natural de acceso directo (autor.nombre = "Nuevo Nombre").

DESAFIO N°55: Crea una función que tome un objeto Autor y devuelva una lista de todos los títulos de libros escritos por el autor. Asegúrate de que la lista de libros sea accesible solo a través de métodos de la clase Autor.

```python
class Libro:
    def __init__(self, titulo: str, isbn: str):
        self._titulo = titulo
        self._isbn = isbn

    @property
    def titulo(self) -> str:
        return self._titulo

    @property
    def isbn(self) -> str:
        return self._isbn


class Autor:
    def __init__(self, nombre: str, nacionalidad: str):
        self._nombre = nombre
        self._nacionalidad = nacionalidad
        self._libros = []  # Lista privada/protegida

    @property
    def nombre(self) -> str:
        return self._nombre

    @property
    def nacionalidad(self) -> str:
        return self._nacionalidad

    @property
    def libros(self) -> list:
        # El acceso a la lista se restringe devolviendo una copia de solo lectura
        return self._libros.copy()

    def agregar_libro(self, libro: Libro) -> None:
        if isinstance(libro, Libro):
            self._libros.append(libro)
        else:
            raise TypeError("El objeto debe ser una instancia de la clase Libro.")


# --- Función solicitada ---
def obtener_titulos_de_autor(autor: Autor) -> list[str]:
    """
    Recibe un objeto Autor y devuelve una lista con los títulos de sus libros.
    Accede a la información únicamente mediante la propiedad/método público de la clase.
    """
    if not isinstance(autor, Autor):
        raise TypeError("El parámetro proporcionado debe ser una instancia de Autor.")

    # Acceso controlado a los libros a través del getter/property de Autor
    return [libro.titulo for libro in autor.libros]


# Ejemplo de uso:
if __name__ == "__main__":
    # Crear autor
    autor_borges = Autor("Jorge Luis Borges", "Argentina")

    # Crear y agregar libros
    autor_borges.agregar_libro(Libro("Ficciones", "978-8420633121"))
    autor_borges.agregar_libro(Libro("El Aleph", "978-8420633114"))
    autor_borges.agregar_libro(Libro("El hacedor", "978-8420633138"))

    # Llamar a la función externa
    titulos = obtener_titulos_de_autor(autor_borges)

    print(f"Lista de títulos de {autor_borges.nombre}:")
    print(titulos)
```

En este caso, el atributo _libros de la clase Autor se mantiene completamente encapsulado y solo se expone a través de la propiedad @property (o su getter). La función externa obtener_titulos_de_autor recibe el objeto Autor y utiliza el acceso público controlado para extraer únicamente los títulos de cada Libro.

DESAFIO N°56: Desafío 56:

Desarrolla una función que reciba una lista de objetos Autor y devuelva el autor que ha escrito el mayor número de libros. Utiliza el encapsulamiento para acceder a la información necesaria de cada objeto Autor.

```python
class Libro:
    def __init__(self, titulo: str, isbn: str):
        self._titulo = titulo
        self._isbn = isbn

    @property
    def titulo(self) -> str:
        return self._titulo


class Autor:
    def __init__(self, nombre: str, nacionalidad: str):
        self._nombre = nombre
        self._nacionalidad = nacionalidad
        self._libros = []

    @property
    def nombre(self) -> str:
        return self._nombre

    @property
    def libros(self) -> list:
        # Devuelve una copia para respetar el encapsulamiento
        return self._libros.copy()

    @property
    def total_libros(self) -> int:
        """Propiedad helper que expone la cantidad de libros de forma encapsulada."""
        return len(self._libros)

    def agregar_libro(self, libro: Libro) -> None:
        if isinstance(libro, Libro):
            self._libros.append(libro)
        else:
            raise TypeError("El objeto debe ser una instancia de la clase Libro.")


# --- Función solicitada ---
def obtener_autor_mas_prolifico(autores: list[Autor]) -> Autor | None:
    """
    Recibe una lista de objetos Autor y devuelve el objeto Autor con mayor número de libros.
    Accede a la información respetando el encapsulamiento de cada objeto.
    """
    if not autores:
        return None

    # Utiliza la función built-in 'max' evaluando la cantidad de libros mediante la propiedad encapsulada
    return max(autores, key=lambda autor: autor.total_libros)


# Ejemplo de uso:
if __name__ == "__main__":
    # Autor 1
    autor1 = Autor("Isabel Allende", "Chilena")
    autor1.agregar_libro(Libro("La casa de los espíritus", "978-8408168270"))
    autor1.agregar_libro(Libro("Largo pétalo de mar", "978-8408224525"))

    # Autor 2
    autor2 = Autor("Stephen King", "Estadounidense")
    autor2.agregar_libro(Libro("El resplandor", "978-0307743657"))
    autor2.agregar_libro(Libro("It", "978-1501142970"))
    autor2.agregar_libro(Libro("Carrie", "978-0307743664"))
    autor2.agregar_libro(Libro("Misery", "978-1501143106"))

    # Autor 3
    autor3 = Autor("Mario Vargas Llosa", "Peruana")
    autor3.agregar_libro(Libro("La ciudad y los perros", "978-8420471839"))

    lista_autores = [autor1, autor2, autor3]

    # Ejecución de la función
    mas_prolifico = obtener_autor_mas_prolifico(lista_autores)

    if mas_prolifico:
        print(f"El autor con más libros es: {mas_prolifico.nombre}")
        print(f"Total de libros registrados: {mas_prolifico.total_libros}")
```
