Desafío 96: Inventario de productos con diccionarios
Desarrolla un sistema de inventario para una tienda utilizando diccionarios, donde:
- La clave es el código del producto.
- El valor es otro diccionario con detalles como nombre, precio y cantidad en stock.
El programa debe permitir:
- Agregar nuevos productos.
- Actualizar el stock de productos existentes.
- Buscar productos por código y mostrar sus detalles.

```python
# Definimos nuestro inventario inicial (diccionario anidado)
inventario = {
    "P001": {"nombre": "Laptop Gamer", "precio": 1200.50, "stock": 5},
    "P002": {"nombre": "Mouse Inalámbrico", "precio": 25.99, "stock": 30}
}

print("=== 1. ESTADO INICIAL DEL INVENTARIO ===")
for codigo, prod in inventario.items():
    print(f"[{codigo}] {prod['nombre']} | Precio: ${prod['precio']:.2f} | Stock: {prod['stock']} uds.")

# -------------------------------------------------------------
# SIMULACIÓN: Agregar un nuevo producto (P003)
# -------------------------------------------------------------
print("\n=== 2. AGREGANDO UN NUEVO PRODUCTO (P003) ===")
nuevo_codigo = "P003"
inventario[nuevo_codigo] = {
    "nombre": "Teclado Mecánico",
    "precio": 75.00,
    "stock": 12
}
print(f"✅ ¡Producto '{inventario[nuevo_codigo]['nombre']}' agregado con el código {nuevo_codigo}!")

# -------------------------------------------------------------
# SIMULACIÓN: Actualizar el stock de un producto existente (P001)
# -------------------------------------------------------------
print("\n=== 3. ACTUALIZANDO STOCK DE LA LAPTOP (P001) ===")
print(f"Stock anterior: {inventario['P001']['stock']} unidades.")
inventario["P001"]["stock"] = 8  # Cambiamos el stock a 8
print(f"✅ ¡Stock actualizado! Nuevo stock de '{inventario['P001']['nombre']}': {inventario['P001']['stock']} unidades.")

# -------------------------------------------------------------
# SIMULACIÓN: Buscar un producto por código (P002)
# -------------------------------------------------------------
print("\n=== 4. BUSCANDO PRODUCTO POR CÓDIGO (P002) ===")
codigo_a_buscar = "P002"
if codigo_a_buscar in inventario:
    prod = inventario[codigo_a_buscar]
    print(f"📦 ¡Encontrado!")
    print(f"   - Nombre: {prod['nombre']}")
    print(f"   - Precio: ${prod['precio']:.2f}")
    print(f"   - Stock:  {prod['stock']} unidades")

# -------------------------------------------------------------
# RESULTADO FINAL
# -------------------------------------------------------------
print("\n=== 5. INVENTARIO FINAL ACTUALIZADO ===")
for codigo, prod in inventario.items():
    print(f"[{codigo}] {prod['nombre']} | Precio: ${prod['precio']:.2f} | Stock: {prod['stock']} uds.")
```

Este ejercicio demuestra la utilidad de los diccionarios anidados (un diccionario dentro de otro) en Python para modelar situaciones del mundo real. Los puntos clave de aprendizaje son:

Estructura de datos avanzada: organizar información compleja relacionando una clave principal única (el código del producto) con un conjunto de atributos detallados (nombre, precio, stock).

Búsqueda y actualización eficiente: A diferencia de las listas, donde a veces necesitas recorrer elemento por elemento, los diccionarios permiten acceder, modificar o buscar datos de forma directa e instantánea a través de sus claves.

Base para sistemas reales: Este enfoque es el fundamento de cómo funcionan los sistemas de gestión de datos, las bases de datos relacionales y el intercambio de información en formato JSON en el desarrollo de software y aplicaciones web.
