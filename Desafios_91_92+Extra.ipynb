Desafío 91: Generación de números de punto flotante
Imagina que estás trabajando en un proyecto de simulación climática y necesitas generar temperaturas aleatorias para un conjunto de datos. Para este propósito, puedes utilizar el método uniform(a, b) de la biblioteca random, que genera un número de punto flotante aleatorio dentro de un rango especificado [a, b]. En este caso, se te solicita generar temperaturas en el rango de -10.0 a 40.0 grados Celsius para simular variaciones climáticas. Estos valores se usarán para análisis posteriores y deberán almacenarse en una estructura adecuada para su manejo y procesamiento. Para python

```python
import random

# Definir el rango de temperaturas (en grados Celsius)
temp_min = -10.0
temp_max = 40.0
num_muestras = 100  # Cantidad de datos a simular

# Generar la lista de temperaturas aleatorias con uniform(a, b)
temperaturas = [random.uniform(temp_min, temp_max) for _ in range(num_muestras)]

# Opcional: Redondear los valores a 2 decimales para mayor legibilidad
temperaturas = [round(t, 2) for t in temperaturas]

# Mostrar los primeros 5 resultados como ejemplo
print("Primeras 5 temperaturas simuladas (Estándar):", temperaturas[:5])
```

Uso de la biblioteca estándar random:
Se combina la función random.uniform() (para generar números decimales aleatorios dentro de un rango específico) con una comprensión de lista (list comprehension), que es la forma más limpia, idiomática y eficiente en Python para crear listas en una sola línea de código.
Se utiliza cuando tu proyecto o simulación es sencillo, pero tambien cuando quieres evitar dependencias externas adicionales y aprovechar al máximo las herramientas que ya vienen integradas en la biblioteca estándar de Python.

Desafío 92: Selección de múltiples elementos únicos



Imagina que estás organizando un sorteo en el que necesitas seleccionar varios ganadores de un grupo grande de participantes. Para este propósito, el método `sample(population, k)` de la biblioteca `random` es una herramienta útil, ya que permite seleccionar de manera aleatoria y única un número determinado de elementos de una población grande. Los ganadores seleccionados serán únicos y representarán una muestra aleatoria del conjunto total de participantes, asegurando que ningún participante pueda ganar más de una vez.

```python
import random

# Lista de participantes en el sorteo
participantes = [
    "Ana",
    "Bruno",
    "Carla",
    "Daniel",
    "Elena",
    "Federico",
    "Gabriela",
    "Hugo",
]

# Cantidad de ganadores que necesitamos seleccionar
cantidad_ganadores = 3

# Selección aleatoria y única de ganadores
ganadores = random.sample(participantes, cantidad_ganadores)

# Mostrar resultados
print("¡Los ganadores del sorteo son:")
for i, ganador in enumerate(ganadores, 1):
    print(f"{i}. {ganador}")
```

El método random.sample(population, k) es una herramienta de la biblioteca estándar de Python cuando se necesita realizar selecciones aleatorias sin reemplazo (es decir, elementos únicos donde nadie puede salir seleccionado más de una vez).

Desafío extra: Simulación de lanzamiento de dados



Imagina que estás creando un juego de mesa virtual y necesitas simular el lanzamiento de dos dados. Para este propósito, puedes utilizar el método `choices(population, weights=k, k=n)` de la biblioteca `random`, que permite seleccionar varios elementos de una población con o sin pesos asociados. En el caso de los dados, si no especificas los pesos, la selección será uniforme, simulando así un lanzamiento justo. Esta técnica te permitirá generar resultados aleatorios de los dados, esenciales para la dinámica del juego.

```python
import random

# Definimos la población: las caras posibles de un dado tradicional (del 1 al 6)
caras_dado = [1, 2, 3, 4, 5, 6]

# Simulamos el lanzamiento de 2 dados (k = 2)
# Al no especificar 'weights', cada cara tiene la misma probabilidad (lanzamiento justo).
lanzamiento = random.choices(caras_dado, k=2)

# Mostramos los resultados individuales y la suma total
dado_1, dado_2 = lanzamiento
suma_total = dado_1 + dado_2

print(f"🎲 Resultado del Dado 1: {dado_1}")
print(f"🎲 Resultado del Dado 2: {dado_2}")
print(f"✨ Suma total del lanzamiento: {suma_total}")
```

Permite repetición (con reemplazo): Si usaras random.sample(), al pedir 2 elementos de una lista de 6, Python te aseguraría que los dos números sean distintos. Sin embargo, en los dados sí pueden salir dos números iguales, lo cual random.choices() maneja perfectamente.

Soporte para dados cargados (pesos): Si en algún momento del juego de mesa quisieras simular un dado tramposo o desequilibrado, podrías pasar el parámetro weights. Por ejemplo, weights=[1, 1, 1, 1, 1, 5] haría que el número 6 salga con mucha más frecuencia
