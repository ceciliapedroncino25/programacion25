Desafío 76:
Construye un árbol binario simple con al menos 3 niveles de profundidad. Cada nodo debe contener un número entero como valor. Una vez construido el árbol, implementa una función que imprima los valores de los nodos siguiendo un recorrido en preorden.

```python
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.izquierdo = None
        self.derecho = None


# 1. Construcción del árbol binario (3 niveles de profundidad)
# Nivel 1 (Raíz)
raiz = Nodo(10)

# Nivel 2 (Hijos de la raíz)
raiz.izquierdo = Nodo(5)
raiz.derecho = Nodo(15)

# Nivel 3 (Nietos / Hijos del nivel 2)
raiz.izquierdo.izquierdo = Nodo(3)
raiz.izquierdo.derecho = Nodo(7)
raiz.derecho.izquierdo = Nodo(12)
raiz.derecho.derecho = Nodo(20)


# 2. Función de recorrido en preorden
def preorden(nodo):
    if nodo is not None:
        print(nodo.valor, end=" ")  # Visitar la raíz
        preorden(nodo.izquierdo)  # Recorrer el subárbol izquierdo
        preorden(nodo.derecho)  # Recorrer el subárbol derecho


# 3. Ejecución del programa
print("Estructura del árbol creada con éxito (3 niveles).")
print("Recorrido en preorden:")
preorden(raiz)
print()
```

¿Cómo funciona el recorrido en preorden?
Para este árbol específico:

Nivel 1: 10 (Raíz)

Nivel 2: 5 (izq), 15 (der)

Nivel 3: 3 y 7 (hijos de 5), 12 y 20 (hijos de 15)

El algoritmo sigue este flujo:

Imprime la raíz principal (10).

Se desplaza completamente por la rama izquierda visitando primero a 5, luego al hijo izquierdo de este (3), y luego al hijo derecho (7).

Regresa y se desplaza por la rama derecha visitando primero a 15, luego a su hijo izquierdo (12), y finalmente a su hijo derecho (20).

Desafío 77:

Dado un árbol binario con números enteros en cada nodo, implementa una función que recorra el árbol en inorden y calcule la suma de todos los valores almacenados en los nodos. El programa debe devolver el resultado final de la suma.

```python
class Nodo:

  def __init__(self, valor):
    self.valor = valor
    self.izquierdo = None
    self.derecho = None


def suma_inorden(nodo):
  if nodo is None:
    return 0
  # Recorrido en inorden: Izquierdo -> Raíz -> Derecho
  suma_izq = suma_inorden(nodo.izquierdo)
  suma_der = suma_inorden(nodo.derecho)

  return nodo.valor + suma_izq + suma_der


# --- Prueba ---
raiz = Nodo(10)
raiz.izquierdo = Nodo(5)
raiz.derecho = Nodo(15)
raiz.izquierdo.izquierdo = Nodo(3)
raiz.izquierdo.derecho = Nodo(7)

print(
    "Suma total de los valores (inorden):", suma_inorden(raiz)
)  # Salida: 40 (10+5+15+3+7)
```

Desafío 78:

Construye un árbol binario en el que cada nodo contiene un número entero. Implementa una función que recorra el árbol en postorden y devuelva el valor máximo encontrado entre todos los nodos del árbol.

```python
def max_postorden(nodo):
  if nodo is None:
    return float(
        "-inf"
    )  # Retorna un valor extremadamente bajo para casos base

  # Recorrido en postorden: Izquierdo -> Derecho -> Raíz
  max_izq = max_postorden(nodo.izquierdo)
  max_der = max_postorden(nodo.derecho)

  return max(nodo.valor, max_izq, max_der)


# --- Prueba usando el árbol anterior ---
print("Valor máximo en el árbol (postorden):", max_postorden(raiz))  # Salida: 15
```

Desafío 79:

Dado un conjunto de números enteros únicos, construye un árbol de búsqueda binaria (ABB) que inserte los valores de manera que el subárbol izquierdo de cada nodo contenga solo nodos con valores menores, y el subárbol derecho solo nodos con valores mayores. Una vez construido el árbol, implementa una función para buscar un número dado y devuelva si ese número existe en el árbol o no.

```python
class NodoABB:

  def __init__(self, valor):
    self.valor = valor
    self.izquierdo = None
    self.derecho = None


def insertar(nodo, valor):
  if nodo is None:
    return NodoABB(valor)
  if valor < nodo.valor:
    nodo.izquierdo = insertar(nodo.izquierdo, valor)
  elif valor > nodo.valor:
    nodo.derecho = insertar(nodo.derecho, valor)
  return nodo


def buscar(nodo, valor):
  if nodo is None:
    return False
  if nodo.valor == valor:
    return True

  if valor < nodo.valor:
    return buscar(nodo.izquierdo, valor)
  else:
    return buscar(nodo.derecho, valor)


# --- Prueba ---
valores = [20, 10, 30, 5, 15, 25, 35]
raiz_abb = None
for v in valores:
  raiz_abb = insertar(raiz_abb, v)

print("¿Existe el 15 en el ABB?", buscar(raiz_abb, 15))  # True
print("¿Existe el 99 en el ABB?", buscar(raiz_abb, 99))  # False
```


Desafío 80: Evaluación de expresiones Matemáticas con árboles



Los árboles de expresiones son una herramienta poderosa en ciencias de la computación y se utilizan comúnmente para evaluar expresiones matemáticas. En este desafío, te propongo construir y evaluar un árbol de expresiones para una expresión matemática dada.



**Tu tarea es escribir un programa en Python que haga lo siguiente:**



Construir el Árbol de Expresiones: Dada una expresión matemática en forma de cadena, construir el árbol de expresiones correspondiente. Puedes asumir que la expresión está bien formada y solo contiene números enteros, y los operadores +, -, *, /.


Evaluar el Árbol de Expresiones: Una vez construido el árbol, evaluarlo y devolver el resultado de la expresión.


Prueba tu Programa: Utiliza la expresión "5 + 3 * 4" para probar tu programa. El resultado debería ser 17.

Para resolver este desafío utilizando la expresión "5 + 3 * 4", utilizaemos una versión simplificada del algoritmo Shunting-Yard para convertir la expresión a postfijo, construir el árbol binario de expresiones (respetando la precedencia de operadores donde * se evalúa antes que +) y luego evaluarlo recursivamente.

```python
class NodoExpresion:

  def __init__(self, valor):
    self.valor = valor
    self.izquierdo = None
    self.derecho = None


def construir_arbol_expresion(expresion):
  tokens = expresion.split()

  def precedencia(op):
    if op in ("+", "-"):
      return 1
    if op in ("*", "/"):
      return 2
    return 0

  # Conversión de Infijo a Postfijo (RPN)
  salida = []
  operadores = []

  for token in tokens:
    if token.isdigit():
      salida.append(token)
    elif token in ("+", "-", "*", "/"):
      while (
          operadores
          and operadores[-1] in ("+", "-", "*", "/")
          and precedencia(operadores[-1]) >= precedencia(token)
      ):
        salida.append(operadores.pop())
      operadores.append(token)

  while operadores:
    salida.append(operadores.pop())

  # Construcción del árbol desde la notación postfija
  pila_nodos = []
  for token in salida:
    if token.isdigit():
      pila_nodos.append(NodoExpresion(token))
    else:
      nodo_op = NodoExpresion(token)
      nodo_op.derecho = pila_nodos.pop()  # El operando derecho va primero
      nodo_op.izquierdo = pila_nodos.pop()  # El operando izquierdo va después
      pila_nodos.append(nodo_op)

  return pila_nodos[0]


def evaluar_arbol(nodo):
  if nodo is None:
    return 0
  # Si es un nodo hoja (número)
  if nodo.izquierdo is None and nodo.derecho is None:
    return int(nodo.valor)

  val_izq = evaluar_arbol(nodo.izquierdo)
  val_der = evaluar_arbol(nodo.derecho)

  if nodo.valor == "+":
    return val_izq + val_der
  elif nodo.valor == "-":
    return val_izq - val_der
  elif nodo.valor == "*":
    return val_izq * val_der
  elif nodo.valor == "/":
    return val_izq / val_der


# --- Prueba del Desafío 80 ---
expresion_str = "5 + 3 * 4"
arbol_exp = construir_arbol_expresion(expresion_str)
resultado = evaluar_arbol(arbol_exp)

print(f"Expresión: {expresion_str}")
print(f"Resultado de la evaluación: {resultado}")  # Salida esperada: 17
```
