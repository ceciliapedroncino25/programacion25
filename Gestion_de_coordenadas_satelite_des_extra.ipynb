### Desafío extra: Gestión de coordenadas de satélites
Crea un programa que registre las coordenadas de varios satélites en órbita en un momento dado. Las coordenadas deben ser inmutables y representadas mediante tuplas. El programa debe:

Almacenar las coordenadas de al menos cinco satélites.
Mostrar la distancia de cada satélite respecto al origen (0, 0, 0).
Identificar el satélite más cercano y el más lejano al origen.
para python

```python
import math

# 1. Almacenar las coordenadas de al menos cinco satélites usando un diccionario
# Las coordenadas (x, y, z) se representan como tuplas para garantizar su inmutabilidad.
satelites = {
    "Alpha-1": (1200, 3400, 5600),
    "Beta-2": (450, 890, 230),
    "Gamma-3": (7800, 1200, 4300),
    "Delta-4": (3200, 4100, 1500),
    "Epsilon-5": (850, 920, 1100),
}

# Diccionario auxiliar para almacenar las distancias calculadas
distancias = {}

print("=" * 60)
print("  ESTADOS Y DISTANCIAS DE SATÉLITES AL ORIGEN (0, 0, 0)")
print("=" * 60)

# 2. Mostrar la distancia de cada satélite respecto al origen (0, 0, 0)
for nombre, coord in satelites.items():
    # Desempaquetar la tupla de coordenadas
    x, y, z = coord

    # Calcular la distancia euclidiana en 3D: sqrt(x^2 + y^2 + z^2)
    distancia = math.sqrt(x**2 + y**2 + z**2)
    distancias[nombre] = distancia

    print(
        f"• Satélite: {nombre:<10} | Coordenadas: {str(coord):<20} | Distancia: {distancia:10.2f} km"
    )

print("-" * 60)

# 3. Identificar el satélite más cercano y el más lejano al origen
# Utilizamos min() y max() evaluando los valores (distancias) del diccionario
mas_cercano = min(distancias, key=distancias.get)
mas_lejano = max(distancias, key=distancias.get)

print("\n--- RESUMEN DE UBICACIÓN ESPACIAL ---")
print(
    f"🌟 **Satélite más cercano:** {mas_cercano} (a {distancias[mas_cercano]:.2f} km)"
)
print(
    f"🌌 **Satélite más lejano:**  {mas_lejano} (a {distancias[mas_lejano]:.2f} km)"
)
print("=" * 60)
```

En esta actividad de desafio se trabaja con los conceptos de:
**Tuplas (x, y, z):** Aseguran que las coordenadas sean inmutables (no se pueden modificar accidentalmente una vez asignadas), lo cual es ideal para representar posiciones fijas en un momento determinado.
**Diccionarios:** Permiten relacionar de manera directa la identidad de cada satélite (su nombre en clave) con su respectiva tupla de coordenadas.
**Módulo math (math.sqrt):** Facilita el cálculo matemático de la distancia euclidiana tridimensional desde el punto $(0,0,0)$.
**Funciones min() y max() con key=distancias.get:** Buscan de forma elegante la clave del diccionario que contiene el valor numérico menor y mayor, respectivamente.
