Desafio N°34 de MODULOS

```python
import string

def procesador_de_texto(texto):
    print(f"--- Texto original ---\n'{texto}'\n")

    # 1. string.capwords(): Formatea el texto tipo Título
    texto_titulo = string.capwords(texto)
    print(f"1. Texto formateado (capwords):\n{texto_titulo}\n")

    # 2. string.punctuation: Elimina signos de puntuación para limpiar el texto
    texto_limpio = "".join(caracter for caracter in texto if caracter not in string.punctuation)
    print(f"2. Texto sin puntuación (punctuation):\n{texto_limpio}\n")

    # 3. string.digits: Extrae únicamente los números presentes en el texto
    numeros_encontrados = "".join(caracter for caracter in texto if caracter in string.digits)
    print(f"3. Números extraídos (digits):\n{numeros_encontrados}\n")

# Ejemplo de uso
texto_ejemplo = "¡hola, mundo! este es el pedido N° 4589 para el cliente."
procesador_de_texto(texto_ejemplo)
```

otro ejemplo del desafio 34: una herramienta de analizador y limpiador de textos que es útil para un procesador de datos usando tres funciones/herramientas diferentes del módulo string:

```python
import string

def analizador_procesador_texto(texto_usuario, usuario):
    print("=== PROCESADOR DE TEXTO: MÓDULO STRING ===\n")

    # 1. string.ascii_letters: Contamos cuántas letras reales tiene el texto
    solo_letras = [c for c in texto_usuario if c in string.ascii_letters]
    cantidad_letras = len(solo_letras)

    # 2. string.whitespace: Reemplazamos tabulaciones o saltos de línea por espacios simples
    texto_limpio = texto_usuario
    for espacio in string.whitespace:
        texto_limpio = texto_limpio.replace(espacio, " ")
    # Eliminamos espacios dobles repetidos
    texto_limpio = " ".join(texto_limpio.split())

    # 3. string.Template: Creamos un reporte con formato profesional usando plantillas
    plantilla_reporte = string.Template(
        "REPORTE GENERADO PARA: $nombre\n"
        "----------------------------------------\n"
        "Texto normalizado: \"$texto\"\n"
        "Total de letras (A-Z, a-z): $letras\n"
    )

    reporte_final = plantilla_reporte.substitute(
        nombre=usuario,
        texto=texto_limpio,
        letras=cantidad_letras
    )

    print(reporte_final)

# --- Ejemplo de prueba ---
texto_sucio = "  Hola   compañeros,\n\t este  es un   texto con   saltos de linea y   espacios!! 123 "
analizador_procesador_texto(texto_sucio, "María")
```

Desafio N°35 de Modulos

```python
import random
import string

def generar_contrasena(longitud=8):
    # 1. Definimos los conjuntos de caracteres usando el módulo 'string'
    minusculas = string.ascii_lowercase  # 'abcdefghijklmnopqrstuvwxyz'
    mayusculas = string.ascii_uppercase  # 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
    numeros = string.digits              # '0123456789'

    # 2. Garantizamos que la contraseña tenga AL MENOS un carácter de cada tipo
    caracteres_obligatorios = [
        random.choice(minusculas),
        random.choice(mayusculas),
        random.choice(numeros)
    ]

    # 3. Completamos el resto de los caracteres (5 más para llegar a 8) eligiendo de todos los tipos
    todos_los_caracteres = minusculas + mayusculas + numeros
    caracteres_restantes = [random.choice(todos_los_caracteres) for _ in range(longitud - 3)]

    # 4. Unimos todos los caracteres y los mezclamos para que el orden sea completamente aleatorio
    lista_contrasena = caracteres_obligatorios + caracteres_restantes
    random.shuffle(lista_contrasena)

    # 5. Convertimos la lista de caracteres a una sola cadena de texto
    contrasena_final = "".join(lista_contrasena)
    return contrasena_final

# Ejemplo de prueba
nueva_contrasena = generar_contrasena(8)
print(f"Tu contraseña segura generada es: {nueva_contrasena}")
```

```python
# ==========================================
# DESAFÍO 36: Módulo y Formato de Texto
# ==========================================

# Definimos las funciones de formato (Módulo)
class FormatoTexto:
    @staticmethod
    def negrita(texto):
        return f"\033[1m{texto}\033[0m"

    @staticmethod
    def italica(texto):
        return f"\033[3m{texto}\033[0m"

    @staticmethod
    def subrayado(texto):
        return f"\033[4m{texto}\033[0m"

    @staticmethod
    def color_rojo(texto):
        return f"\033[91m{texto}\033[0m"

    @staticmethod
    def color_verde(texto):
        return f"\033[92m{texto}\033[0m"


# --- Programa Principal ---
def ejecutar():
    print("=== DEMOSTRACIÓN DE FORMATO ===\n")

    frase = "Este texto tiene formato dinámico en Python"

    print(FormatoTexto.negrita(f"1. Negrita: {frase}"))
    print(FormatoTexto.italica(f"2. Itálica: {frase}"))
    print(FormatoTexto.subrayado(f"3. Subrayado: {frase}"))
    print(FormatoTexto.color_rojo(f"4. Color Rojo: {frase}"))
    print(FormatoTexto.color_verde(f"5. Color Verde: {frase}"))

if __name__ == "__main__":
    ejecutar()
```

```python
# =========================================================
# DESAFÍO 36: Módulo y Formatos de Texto (Versión Unificada)
# =========================================================

# --- Módulo / Funciones de Formato ---
def estilo_titulo(texto):
    """Convierte el texto a formato Título y lo subraya."""
    texto_formateado = texto.strip().title()
    linea = "=" * len(texto_formateado)
    return f"{texto_formateado}\n{linea}"

def enmarcar(texto, caracter="*"):
    """Encierra una línea de texto dentro de un marco decorativo."""
    longitud = len(texto) + 4
    borde = caracter * longitud
    return f"{borde}\n{caracter} {texto} {caracter}\n{borde}"

def centrar_texto(texto, ancho=50, relleno="-"):
    """Centra el texto completando los lados con el carácter especificado."""
    return texto.center(ancho, relleno)

def alternar_mayusculas(texto):
    """Alterna entre minúsculas y mayúsculas en el texto."""
    resultado = []
    for i, char in enumerate(texto):
        if i % 2 == 0:
            resultado.append(char.upper())
        else:
            resultado.append(char.lower())
    return "".join(resultado)


# --- Programa Principal ---
def ejecutar_demostracion():
    print("=== DEMOSTRACIÓN DE FORMATOS DE TEXTO ===\n")

    # 1. Aplicando formato de Título
    print(estilo_titulo("bienvenidos al procesador de texto"))
    print()

    # 2. Aplicando marco decorativo
    print(enmarcar("AVISO IMPORTANTE", caracter="#"))
    print()

    # 3. Centrando un texto
    print(centrar_texto(" SECCIÓN DE CÓDIGO ", ancho=40, relleno="*"))
    print()

    # 4. Alternando mayúsculas y minúsculas
    frase = "programacion en python"
    print(f"Texto alternado: {alternar_mayusculas(frase)}")

if __name__ == "__main__":
    ejecutar_demostracion()
```

Desafio N°37 de Modulos

```python
from collections import Counter
import re
import matplotlib.pyplot as plt

def analizar_y_graficar_texto(texto, top_n=10):
    # 1. Limpieza del texto: Convertir a minúsculas y quitar signos de puntuación usando expresiones regulares
    texto_limpio = re.sub(r'[^\w\s]', '', texto.lower())

    # Separar en palabras individuales
    palabras = texto_limpio.split()

    # Opcional: Filtro de "stop words" (palabras cortas como conectores/artículos que no aportan mucho significado)
    palabras_filtradas = [p for p in palabras if len(p) > 2]

    # 2. Utilizar Counter del módulo collections para contar frecuencias
    contador_palabras = Counter(palabras_filtradas)

    # Obtener las N palabras más comunes
    mas_comunes = contador_palabras.most_common(top_n)

    # Mostrar resultados en consola
    print(f"=== LAS {top_n} PALABRAS MÁS FRECUENTES ===")
    for palabra, frecuencia in mas_comunes:
        print(f"• '{palabra}': {frecuencia} veces")

    # 3. Preparar datos para el gráfico de barras
    palabras_grafico = [item[0] for item in mas_comunes]
    frecuencias_grafico = [item[1] for item in mas_comunes]

    # 4. Crear el gráfico con Matplotlib
    plt.figure(figsize=(10, 6))

    # Dibujar barras verticales (puedes cambiar 'skyblue' por el color que prefieras)
    barras = plt.bar(palabras_grafico, frecuencias_grafico, color='skyblue', edgecolor='navy')

    # Configuración de títulos y etiquetas
    plt.title(f"Top {top_n} Palabras más Frecuentes", fontsize=14, fontweight='bold')
    plt.xlabel("Palabras", fontsize=12)
    plt.ylabel("Frecuencia", fontsize=12)
    plt.grid(axis='y', linestyle='--', alpha=0.7)

    # Agregar el valor exacto encima de cada barra
    for barra in barras:
        yval = barra.get_height()
        plt.text(barra.get_x() + barra.get_width()/2.0, yval + 0.1, int(yval), ha='center', va='bottom', fontweight='bold')

    # Mostrar la ventana del gráfico
    plt.tight_layout()
    plt.show()

# --- Texto de prueba ---
texto_ejemplo = """
Python es un lenguaje de programación muy popular. Python es fácil de aprender y Python es muy potente.
La programación en Python permite desarrollo web, análisis de datos, inteligencia artificial y automatización.
Aprender programación con Python abre muchas puertas en el mundo de la tecnología.
"""

# Ejecutamos la función
analizar_y_graficar_texto(texto_ejemplo, top_n=10)
```

Desafio N°38 de MODULOS

```python
import os

def guardar_archivo(nombre_archivo, contenido):
    # Aseguramos que termine en .txt
    if not nombre_archivo.endswith(".txt"):
        nombre_archivo += ".txt"

    # Guardamos el archivo con codificacion utf-8
    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        archivo.write(contenido)

    # os.path.abspath obtiene la ruta absoluta en el sistema
    ruta = os.path.abspath(nombre_archivo)
    print(f"\n[ÉXITO] Archivo guardado correctamente en:\n{ruta}")

def leer_archivo(nombre_archivo):
    if not nombre_archivo.endswith(".txt"):
        nombre_archivo += ".txt"

    # os.path.exists comprueba si el archivo existe en el disco
    if not os.path.exists(nombre_archivo):
        print(f"\n[ERROR] El archivo '{nombre_archivo}' no existe.")
        return

    with open(nombre_archivo, "r", encoding="utf-8") as archivo:
        contenido = archivo.read()

    print(f"\n--- Contenido de '{nombre_archivo}' ---")
    print(contenido)
    print("---------------------------------------")

def listar_archivos_txt():
    # os.getcwd() obtiene la carpeta actual donde trabajas
    carpeta_actual = os.getcwd()
    # os.listdir() lista el contenido de la carpeta
    archivos = os.listdir(carpeta_actual)

    # Filtramos los archivos que terminan en .txt
    archivos_txt = [arch for arch in archivos if arch.endswith(".txt")]

    print(f"\nArchivos .txt en la carpeta ({carpeta_actual}):")
    if archivos_txt:
        for i, arch in enumerate(archivos_txt, 1):
            print(f"  {i}. {arch}")
    else:
        print("  (No hay archivos .txt guardados aún)")

# --- Prueba automática para verificar que funciona ---
def probar_desafío38():
    print("=== DESAFÍO 38: MÓDULO OS ===")

    # 1. Guardar un archivo de prueba
    guardar_archivo("mi_documento.txt", "Hola! Este archivo fue guardado usando el modulo os de Python.")

    # 2. Listar los archivos .txt de la carpeta
    listar_archivos_txt()

    # 3. Leer el archivo creado
    leer_archivo("mi_documento.txt")

if __name__ == "__main__":
    probar_desafío38()
```

¿Por qué suele dar error el código con os?
Error FileNotFoundError: Ocurre si intentas leer un archivo antes de crearlo o si escribes mal el nombre (por ejemplo, escribir notas en lugar de notas.txt). En el código corregido de arriba, el programa añade automáticamente el .txt si se te olvida.

Uso de menú interactivo input() en algunas plataformas: Si estás ejecutando el código dentro de un entorno interactivo que no soporta entradas de consola directas, el input() puede fallar. En el script de arriba se usa una prueba automática que crea, lista y lee el archivo sin detenerse a pedir entradas.

Desafio N°39 de MODULOS

```python
from datetime import datetime

def calcular_diferencia_fechas(fecha_str1, fecha_str2):
    """
    Recibe dos cadenas de texto en formato 'DD/MM/YYYY'
    y calcula la diferencia en días entre ambas fechas.
    """
    formato = "%d/%m/%Y"

    # 1. Convertir cadenas de texto (str) a objetos datetime (strptime)
    fecha1 = datetime.strptime(fecha_str1, formato)
    fecha2 = datetime.strptime(fecha_str2, formato)

    # 2. Calcular la diferencia (se obtiene un objeto timedelta)
    diferencia = abs(fecha2 - fecha1) # abs() garantiza un número positivo de días

    return diferencia.days

def programa_calculadora_fechas():
    print("=== CALCULADORA DE FECHAS (MÓDULO DATETIME) ===\n")

    # Ejemplos de prueba directos
    f1 = "01/01/2026"
    f2 = "13/08/2026"

    dias = calcular_diferencia_fechas(f1, f2)

    print(f"Fecha inicial : {f1}")
    print(f"Fecha final   : {f2}")
    print(f"---------------------------------------")
    print(f"Diferencia    : {dias} días transcurridos")

if __name__ == "__main__":
    programa_calculadora_fechas()
```

Estructura del Código:
datetime.strptime(): Convierte entradas en formato de fecha estandarizado DD/MM/YYYY.

Resta de fechas: Genera una instancia de timedelta con la diferencia calculada en días.
