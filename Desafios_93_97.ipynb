 Desafío 93: Traductor multilingüe mejorado
Utilizando diccionarios, extiende el ejemplo de traducciones para:

Incluir más palabras y más idiomas (por ejemplo, italiano "it", alemán "de").
Implementar una función que, dada una palabra en inglés y una lista de idiomas, devuelva sus traducciones en esos idiomas.
Manejar el caso en que una traducción no esté disponible, indicando al usuario que no se encontró.

```python
# 1. Diccionario anidado con palabras en inglés y sus traducciones en varios idiomas
# Clave principal: palabra en inglés
# Subdiccionario: código de idioma ('es', 'fr', 'it', 'de') y la traducción
traducciones_db = {
    "hello": {
        "es": "hola",
        "fr": "bonjour",
        "it": "ciao",
        "de": "hallo"
    },
    "apple": {
        "es": "manzana",
        "fr": "pomme",
        "it": "mela",
        "de": "apfel"
    },
    "dog": {
        "es": "perro",
        "fr": "chien",
        "it": "cane"
        # Nota: 'de' (alemán) no está disponible a propósito para probar el manejo de errores
    },
    "book": {
        "es": "libro",
        "fr": "livre",
        "it": "libro",
        "de": "buch"
    },
    "sun": {
        "es": "sol",
        "fr": "soleil",
        "it": "sole",
        "de": "sonne"
    }
}

# 2. Función que dada una palabra y una lista de idiomas devuelve las traducciones
def traducir_palabra(palabra, lista_idiomas):
    """
    Busca una palabra en inglés en la base de datos de traducciones
    para los idiomas especificados en la lista.
    """
    palabra_lower = palabra.lower().strip()
    resultados = {}

    # Verificar si la palabra en inglés existe en nuestro diccionario principal
    if palabra_lower not in traducciones_db:
        print(f"⚠️ Aviso: La palabra '{palabra}' no se encuentra en la base de datos del traductor.")
        return None

    # Obtener el diccionario de traducciones para esa palabra
    idiomas_disponibles = traducciones_db[palabra_lower]

    print(f"\n--- Traducciones para: '{palabra_lower.upper()}' ---")

    for idioma in lista_idiomas:
        idioma_clean = idioma.lower().strip()

        # Manejar el caso en que el idioma o la traducción específica no estén disponibles
        if idioma_clean in idiomas_disponibles:
            traduccion = idiomas_disponibles[idioma_clean]
            resultados[idioma_clean] = traduccion
            print(f"  [{idioma_clean.upper()}] -> {traduccion}")
        else:
            resultados[idioma_clean] = None
            print(f"  [{idioma_clean.upper()}] -> ❌ Traducción no disponible")

    return resultados


# ==========================================
# 3. Pruebas / Casos de uso del programa
# ==========================================

# Caso 1: Palabra existente y todos los idiomas disponibles
idiomas_solicitados_1 = ["es", "fr", "it", "de"]
traducir_palabra("apple", idiomas_solicitados_1)

# Caso 2: Palabra existente pero falta un idioma en la base de datos (ej. "dog" no tiene alemán "de")
idiomas_solicitados_2 = ["es", "it", "de"]
traducir_palabra("dog", idiomas_solicitados_2)

# Caso 3: Palabra que NO existe en la base de datos
idiomas_solicitados_3 = ["es", "fr"]
traducir_palabra("computer", idiomas_solicitados_3)
```

Para este desafio se utiliza:
Estructura de datos avanzada: Usar diccionarios anidados permite escalar fácilmente el programa añadiendo nuevos términos o idiomas sin alterar la lógica principal.
Validación de existencia (in): Comprobamos tanto si la palabra en inglés existe en el diccionario principal como si el código del idioma solicitado se encuentra dentro de sus traducciones.
Manejo de casos faltantes: Si un idioma no está registrado para una palabra específica, el programa informa elegantemente al usuario (❌ Traducción no disponible) en lugar de mostrar un error de tipo KeyError.

Desafío 94: Gestión de usuarios en concursos
Tienes tres listas de usuarios de @1001problemas.com que han participado en los concursos "AStronomía", "Matemática" y "Programación". Utiliza conjuntos para:

Encontrar los usuarios que han participado en todos los concursos.
Encontrar los usuarios que solo han participado en "Programación".
Listar todos los usuarios únicos que han participado en al menos un concurso.

```python
# 1. Definir los conjuntos con los correos de los participantes de cada concurso
astronomia = {
    "ana@1001problemas.com",
    "carlos@1001problemas.com",
    "lucia@1001problemas.com",
    "pedro@1001problemas.com",
}

matematica = {
    "carlos@1001problemas.com",
    "lucia@1001problemas.com",
    "sofia@1001problemas.com",
    "mateo@1001problemas.com",
}

programacion = {
    "carlos@1001problemas.com",
    "lucia@1001problemas.com",
    "pedro@1001problemas.com",
    "juan@1001problemas.com",
}

print("=" * 65)
print("  REPORTE DE PARTICIPACIÓN EN CONCURSOS DE @1001problemas.com")
print("=" * 65)

# -------------------------------------------------------------
# 1. Usuarios que han participado en todos los concursos
# -------------------------------------------------------------
# Utilizamos el operador de intersección (&) para encontrar los elementos comunes en los tres conjuntos.
todos_los_concursos = astronomia & matematica & programacion

print("\n🌟 1. Usuarios que han participado en TODOS los concursos:")
if todos_los_concursos:
    for usuario in todos_los_concursos:
        print(f"   • {usuario}")
else:
    print("   (Ningún usuario participó en los tres concursos)")

# -------------------------------------------------------------
# 2. Usuarios que SOLO han participado en "Programación"
# -------------------------------------------------------------
# Restamos a 'programacion' la unión de los otros dos concursos para excluir
# a quienes también participaron en astronomía o matemática.
solo_programacion = programacion - (astronomia | matematica)

print("\n💻 2. Usuarios que SOLO han participado en 'Programación':")
if solo_programacion:
    for usuario in solo_programacion:
        print(f"   • {usuario}")
else:
    print("   (No hay usuarios exclusivos de programación)")

# -------------------------------------------------------------
# 3. Todos los usuarios únicos que han participado en al menos un concurso
# -------------------------------------------------------------
# Utilizamos el operador de unión (|) para combinar todos los conjuntos sin duplicados.
usuarios_unicos = astronomia | matematica | programacion

print(
    f"\n👥 3. Total de usuarios únicos ({len(usuarios_unicos)} en total) que participaron en al menos un concurso:"
)
for usuario in sorted(usuarios_unicos):  # sorted() los ordena alfabéticamente
    print(f"   • {usuario}")

print("=" * 65)
```

Los operadores utilizados para este desafio fueron los siguientes:
Intersección (& o método .intersection()): Devuelve únicamente los elementos que están presentes en todos los conjuntos evaluados al mismo tiempo (ideal para el punto 1).
Unión (| o método .union()): Combina los elementos de los conjuntos eliminando automáticamente los duplicados (necesario para el punto 3).
Diferencia (- o método .difference()): Permite quitar de un conjunto aquellos elementos que también están presentes en otro(s) (clave en el punto 2 para aislar únicamente a los de programación restando el resto de los participantes).


Desafío 95: Análisis de datos de ventas con tuplas
Una tienda en línea registra cada venta como una tupla que contiene (id_producto, cantidad, precio_unitario). Crea un programa que:

Almacene una lista de ventas realizadas.
Calcule el ingreso total generado.
Identifique el producto más vendido en términos de cantidad.

```python
# 1. Almacenar una lista de ventas realizadas
# Cada tupla representa: (id_producto, cantidad, precio_unitario)
ventas = [
    ("PROD-A", 3, 15.50),
    ("PROD-B", 10, 5.00),
    ("PROD-A", 5, 15.50),
    ("PROD-C", 2, 45.00),
    ("PROD-B", 4, 5.00),
    ("PROD-D", 1, 120.00),
    ("PROD-C", 6, 45.00),
]

print("=" * 60)
print("          REPORTE DE VENTAS DE LA TIENDA")
print("=" * 60)

# Inicializar variables de control
ingreso_total = 0.0
cantidades_por_producto = {}

# Procesar cada venta registrada en la lista
for venta in ventas:
    # Desempaquetar la tupla
    id_producto, cantidad, precio_unitario = venta

    # 2. Calcular el ingreso parcial y sumarlo al total
    ingreso_venta = cantidad * precio_unitario
    ingreso_total += ingreso_venta

    # Acumular la cantidad total vendida por cada producto
    if id_producto in cantidades_por_producto:
        cantidades_por_producto[id_producto] += cantidad
    else:
        cantidades_por_producto[id_producto] = cantidad

    print(f"• Producto: {id_producto:<8} | Cantidad: {cantidad:<3} | Precio U.: ${precio_unitario:<6.2f} | Subtotal: ${ingreso_venta:<7.2f}")

print("-" * 60)
print(f"💵 **Ingreso total generado:** ${ingreso_total:.2f}")

# 3. Identificar el producto más vendido en términos de cantidad
# Buscamos la clave del diccionario cuya cantidad acumulada sea la mayor
producto_mas_vendido = max(cantidades_por_producto, key=cantidades_por_producto.get)
max_cantidad = cantidades_por_producto[producto_mas_vendido]

print(f"📈 **Producto más vendido:** {producto_mas_vendido} (con {max_cantidad} unidades totales)")
print("=" * 60)
```

En este desafio se utilizaron:
Tuplas para registros estructurados: Cada venta (id_producto, cantidad, precio_unitario) funciona como un registro inmutable y ligero, perfecto para representar filas de una base de datos o transacciones.
Desempaquetado de tuplas: En el bucle for, descomponemos directamente la tupla en variables legibles (id_producto, cantidad, precio_unitario), lo que hace el código mucho más limpio.
Agregación con diccionarios: Combinamos la lista de tuplas con un diccionario (cantidades_por_producto) para sumar las cantidades de productos que aparecen en múltiples transacciones (como PROD-A o PROD-B).
Función max() con key personalizada: Permite encontrar el producto con mayor volumen de ventas evaluando los valores numéricos del diccionario de forma directa y elegante.

Desafío 96: Inventario de productos con diccionarios
Desarrolla un sistema de inventario para una tienda utilizando diccionarios, donde:

La clave es el código del producto.
El valor es otro diccionario con detalles como nombre, precio y cantidad en stock.
El programa debe permitir:

Agregar nuevos productos.
Actualizar el stock de productos existentes.
Buscar productos por código y mostrar sus detalles.

```python
# 1. Estructura inicial del inventario (Diccionario anidado)
inventario = {
    "P001": {"nombre": "Laptop Pro 15\"", "precio": 1200.50, "stock": 8},
    "P002": {"nombre": "Mouse Inalámbrico", "precio": 25.99, "stock": 45},
    "P003": {"nombre": "Teclado Mecánico", "precio": 75.00, "stock": 20}
}

# -------------------------------------------------------------
# Funciones del Sistema
# -------------------------------------------------------------

def agregar_producto(inventario, codigo, nombre, precio, stock):
    """Agrega un nuevo producto al inventario si el código no existe."""
    codigo_limpio = codigo.upper().strip()
    if codigo_limpio in inventario:
        print(f"⚠️ El código '{codigo_limpio}' ya está registrado. Use la función de actualizar.")
    else:
        inventario[codigo_limpio] = {
            "nombre": nombre,
            "precio": float(precio),
            "stock": int(stock)
        }
        print(f"✅ Producto '{nombre}' agregado exitosamente con el código {codigo_limpio}.")

def actualizar_stock(inventario, codigo, cantidad):
    """Actualiza el stock de un producto existente (suma o resta unidades)."""
    codigo_limpio = codigo.upper().strip()
    if codigo_limpio in inventario:
        inventario[codigo_limpio]["stock"] += cantidad
        nuevo_stock = inventario[codigo_limpio]["stock"]
        nombre = inventario[codigo_limpio]["nombre"]
        print(f"🔄 Stock actualizado para '{nombre}'. Nuevo stock: {nuevo_stock} unidades.")
    else:
        print(f"❌ Error: El código '{codigo_limpio}' no se encuentra en el inventario.")

def buscar_producto(inventario, codigo):
    """Busca un producto por su código y muestra sus detalles."""
    codigo_limpio = codigo.upper().strip()
    if codigo_limpio in inventario:
        prod = inventario[codigo_limpio]
        print(f"\n" + "="*40)
        print(f" 🔍 DETALLES DEL PRODUCTO [{codigo_limpio}]")
        print("="*40)
        print(f"  • Nombre : {prod['nombre']}")
        print(f"  • Precio : ${prod['precio']:.2f}")
        print(f"  • Stock  : {prod['stock']} unidades")
        print("="*40)
    else:
        print(f"❌ No se encontró ningún producto con el código '{codigo_limpio}'.")


# =============================================================
# Pruebas y Casos de Uso del Sistema
# =============================================================

print("=" * 60)
print("       DEMOSTRACIÓN DEL SISTEMA DE INVENTARIO")
print("=" * 60)

# 1. Buscar un producto existente
buscar_producto(inventario, "P001")

# 2. Agregar un nuevo producto
agregar_producto(inventario, "P004", "Monitor 27 Pulgadas", 310.00, 14)

# 3. Actualizar el stock de un producto (ej. ingresan 5 unidades más de mouse)
actualizar_stock(inventario, "P002", 5)

# 4. Buscar el nuevo producto para verificar que se guardó bien
buscar_producto(inventario, "P004")

# 5. Intentar buscar un producto que no existe (manejo de errores)
buscar_producto(inventario, "P999")
```

Para este desafio se implementaron:
Diccionarios anidados: Permiten estructurar información compleja del estilo clave-valor dentro de otra clave-valor, simulando registros tabulares o filas de una base de datos de manera ligera en memoria.
Búsqueda rápida por clave (in): Python busca elementos en un diccionario utilizando las claves (codigo en este caso) con una eficiencia óptima $O(1)$, haciendo que la consulta sea instantánea sin importar el tamaño del inventario.
Modularidad: El uso de funciones independientes (agregar_producto, actualizar_stock, buscar_producto) mantiene el código limpio, reutilizable y fácil de integrar posteriormente en un menú interactivo con bucles while.

 Desafío 97: Operaciones entre conjuntos de datos
En un estudio de mercado, se tienen tres conjuntos de encuestados según las plataformas que usan: Facebook, Instagram, y Twitter (X). Utiliza conjuntos para:

Encontrar usuarios que usan las tres plataformas.
Usuarios que usan solo una de las plataformas.
Usuarios que usan al menos dos plataformas.

```python
# 1. Definir los conjuntos con los correos de los usuarios por plataforma
facebook = {
    "ana@mail.com", "carlos@mail.com", "lucia@mail.com",
    "pedro@mail.com", "sofia@mail.com"
}

instagram = {
    "carlos@mail.com", "lucia@mail.com", "mateo@mail.com",
    "sofia@mail.com", "elena@mail.com"
}

twitter = {
    "lucia@mail.com", "pedro@mail.com", "mateo@mail.com",
    "elena@mail.com", "juan@mail.com"
}

print("=" * 65)
print("         ESTUDIO DE MERCADO: USO DE PLATAFORMAS")
print("=" * 65)

# -------------------------------------------------------------
# 1. Encontrar usuarios que usan las tres plataformas
# -------------------------------------------------------------
# Utilizamos la intersección múltiple (&) para hallar los elementos comunes en los tres conjuntos.
tres_plataformas = facebook & instagram & twitter

print("\n🌟 1. Usuarios que usan las TRES plataformas:")
if tres_plataformas:
    for usuario in tres_plataformas:
        print(f"   • {usuario}")
else:
    print("   (Ningún usuario utiliza las tres plataformas)")

# -------------------------------------------------------------
# 2. Usuarios que usan solo una de las plataformas
# -------------------------------------------------------------
# A cada conjunto le restamos la unión de los otros dos para aislar a los usuarios exclusivos.
solo_facebook = facebook - (instagram | twitter)
solo_instagram = instagram - (facebook | twitter)
solo_twitter = twitter - (facebook | instagram)

# Unimos los usuarios exclusivos de cada plataforma
solo_una = solo_facebook | solo_instagram | solo_twitter

print("\n📱 2. Usuarios que usan SOLO UNA plataforma:")
if solo_una:
    for usuario in sorted(solo_una):
        print(f"   • {usuario}")
else:
    print("   (No hay usuarios exclusivos de una sola plataforma)")

# -------------------------------------------------------------
# 3. Usuarios que usan al menos dos plataformas
# -------------------------------------------------------------
# Al total de usuarios únicos (unión de todos) le restamos aquellos que usan una sola plataforma,
# obteniendo exactamente los que participan en 2 o en las 3 plataformas.
todos_unicos = facebook | instagram | twitter
al_menos_dos = todos_unicos - solo_una

print("\n🔗 3. Usuarios que usan AL MENOS DOS plataformas:")
if al_menos_dos:
    for usuario in sorted(al_menos_dos):
        print(f"   • {usuario}")
else:
    print("   (Ningún usuario utiliza dos o más plataformas)")

print("=" * 65)
```

Para este desafio se utilizaron:
Intersección (&): Extrae de forma eficiente los elementos que se repiten simultáneamente en todos los grupos evaluados (ideal para encontrar a los usuarios presentes en Facebook, Instagram y Twitter a la vez).
Diferencia (-): Permite descontar elementos. Restarle a un conjunto la unión de los demás nos aísla con precisión quirúrgica a los usuarios que solo usan esa red social específica.
Aritmética de conjuntos (Unión total - Solo una): En lugar de evaluar combinaciones complejas de pares, calcular el universo total de usuarios y sustraer a los de uso exclusivo es la forma más elegante y rápida de encontrar a quienes usan al menos dos plataformas.
