Desafio N°40 RECURSIVIDAD.

````python
def mcd_recursivo(a, b):
    """
    Calcula el Máximo Común Divisor (MCD) de dos números enteros a y b
    de forma recursiva utilizando el Algoritmo de Euclides.
    """
    # Caso base: si el divisor es 0, el MCD es 'a'
    if b == 0:
        return a
    # Caso recursivo: el MCD de (a, b) es el MCD de (b, a % b)
    return mcd_recursivo(b, a % b)


# --- Demostración ---
print("=== DESAFÍO 40: MCD RECURSIVO ===")
num1, num2 = 8, 12
resultado = mcd_recursivo(num1, num2)
print(f"El MCD de {num1} y {num2} es: {resultado}")  # Salida: 4
````

````python
def mcd_dos_numeros(a, b):
    if b == 0:
        return a
    return mcd_dos_numeros(b, a % b)

def mcd_lista_recursivo(lista_numeros):
    """
    Calcula el MCD de una lista entera de números de forma recursiva.
    """
    # Caso base 1: Lista vacía
    if not lista_numeros:
        return 0
    # Caso base 2: Si solo queda un número, ese es el MCD
    if len(lista_numeros) == 1:
        return lista_numeros[0]

    # Paso recursivo: MCD del primer número con el MCD del resto de la lista
    return mcd_dos_numeros(lista_numeros[0], mcd_lista_recursivo(lista_numeros[1:]))

# --- Prueba ---
numeros = [24, 36, 60, 48]
resultado_lista = mcd_lista_recursivo(numeros)
print(f"El MCD de la lista {numeros} es: {resultado_lista}")  # Salida: 12
````

Tambien se puede usar la propiedad asociativa del MCD: $\text{MCD}(a, b, c) = \text{MCD}(\text{MCD}(a, b), c)$.

Desafio N°41 de RECURSIVIDAD.

````python
# Simulamos las funciones del módulo personalizado 'formato_texto'
def resaltar_palabra(texto):
    """Devuelve la palabra con formato ANSI (Negrita y Rojo)."""
    return f"\033[1;91m{texto}\033[0m"


def resaltar_ocurrencias_recursivo(palabras, palabra_clave, indice=0):
    """
    Recorre recursivamente una lista de palabras y aplica formato
    a aquellas que coincidan con la palabra clave.
    """
    # Caso base: si alcanzamos el final de la lista, retornamos una lista vacía
    if indice == len(palabras):
        return []

    palabra_actual = palabras[indice]

    # Limpiamos la palabra de puntuación para comparar correctamente
    palabra_limpia = palabra_actual.strip(".,;:¡!¿?").lower()

    if palabra_limpia == palabra_clave.lower():
        palabra_procesada = resaltar_palabra(palabra_actual)
    else:
        palabra_procesada = palabra_actual

    # Paso recursivo: procesamos el siguiente índice y acumulamos el resultado
    return [palabra_procesada] + resaltar_ocurrencias_recursivo(
        palabras, palabra_clave, indice + 1
    )


# --- Demostración ---
print("\n=== DESAFÍO 41: RESALTADO RECURSIVO ===")
texto_largo = "Python es un lenguaje genial. Con Python podemos programar de todo, porque Python es versátil."
palabra_buscar = "Python"

# Dividimos el texto en palabras y ejecutamos la función recursiva
lista_palabras = texto_largo.split()
resultado_lista = resaltar_ocurrencias_recursivo(lista_palabras, palabra_buscar)
texto_final = " ".join(resultado_lista)

print(f"Texto original:\n{texto_largo}\n")
print(f"Texto con '{palabra_buscar}' resaltado:")
print(texto_final)
````

Usamos el módulo formato_texto (o su definición en función) para destacar en negrita y color todas las apariciones de una palabra clave en un texto de forma recursiva.

Desafio N°42 de RECURSIVIDAD

````python
def extraer_palabras(oracion, palabra_actual="", lista_palabras=None):
    """
    Función auxiliar recursiva para dividir una oración en una lista de palabras
    sin utilizar str.split().
    """
    if lista_palabras is None:
        lista_palabras = []

    if not oracion:
        if palabra_actual:
            lista_palabras.append(palabra_actual)
        return lista_palabras

    primer_caracter = oracion[0]
    resto_oracion = oracion[1:]

    if primer_caracter == " ":
        if palabra_actual:
            lista_palabras.append(palabra_actual)
        return extraer_palabras(resto_oracion, "", lista_palabras)
    else:
        return extraer_palabras(
            resto_oracion, palabra_actual + primer_caracter, lista_palabras
        )


def invertir_lista_palabras_recursivo(palabras):
    """Invierte recursivamente el orden de los elementos en una lista."""
    # Caso base: lista vacía
    if not palabras:
        return []

    # Paso recursivo: el último elemento + el resto invertido
    return [palabras[-1]] + invertir_lista_palabras_recursivo(palabras[:-1])


def unir_palabras_recursivo(palabras, separador=" "):
    """Une una lista de palabras en un string recursivamente sin str.join()."""
    if not palabras:
        return ""
    if len(palabras) == 1:
        return palabras[0]
    return palabras[0] + separador + unir_palabras_recursivo(palabras[1:])


def invertir_oracion_recursiva(oracion):
    """Función principal que orquesta la inversión recursiva."""
    palabras = extraer_palabras(oracion)
    palabras_invertidas = invertir_lista_palabras_recursivo(palabras)
    return unir_palabras_recursivo(palabras_invertidas)


# --- Demostración ---
print("\n=== DESAFÍO 42: INVERTIR PALABRAS RECURSIVAMENTE ===")
oracion_original = "Aprender programación con Python es divertido"
oracion_invertida = invertir_oracion_recursiva(oracion_original)

print(f"Oración original : '{oracion_original}'")
print(f"Oración invertida: '{oracion_invertida}'")
````

Para invertir las palabras: se dividen las palabras de la oración de forma manual (buscando espacios) y se reestructuran mediante llamadas recursivas.

Desafio N°43 de RECURSIVIDAD

````python
def suma_hasta_n(n):
    """Calcula la suma acumulada desde 1 hasta n de forma recursiva."""
    if n <= 0:
        return 0
    return n + suma_hasta_n(n - 1)


def mono_cuenta_manzanas(n_manzanas):
    """
    Simula cómo el mono cuenta manzanas, reiniciando el conteo
    desde 1 en cada iteración y acumulando la suma total.
    """
    # Caso base: si no hay manzanas, el total contado es 0
    if n_manzanas <= 0:
        return 0

    # Paso recursivo:
    # 1. El mono cuenta hasta 'n_manzanas' acumulando suma_hasta_n(n_manzanas)
    # 2. Se suma recursivamente el resultado de la pila anterior (n_manzanas - 1)
    conteo_actual = suma_hasta_n(n_manzanas)
    print(f"• Pila hasta {n_manzanas} manzanas: contó {conteo_actual} en esta ronda.")

    return conteo_actual + mono_cuenta_manzanas(n_manzanas - 1)


# --- Demostración ---
print("\n=== DESAFÍO 43: EL MONO QUE CUENTA MANZANAS ===")
pila = 5
total_acumulado = mono_cuenta_manzanas(pila)

print(f"\nTotal de conteos acumulados para {pila} manzanas: {total_acumulado}")
````

En este problema: si el mono tiene $N$ manzanas, cuenta $1$ manzana, luego cuenta $1, 2$, luego $1, 2, 3$, y así sucesivamente hasta llegar a $N$. La suma total acumulada equivale a la suma de los números triangulares:$$\text{Total}(N) = \sum_{i=1}^{N} i = \frac{N \times (N + 1)}{2}$$
