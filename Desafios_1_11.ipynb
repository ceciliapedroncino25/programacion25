Desafío 1: Suma de los primeros $n$ números naturalesPuedes resolverlo mediante un bucle, usando la función integrada sum() o aplicando directamente la fórmula matemática ($n \times (n + 1) / 2$), que es la más eficiente.

```python
def suma_primeros_n(n):
  # Opción eficiente (Fórmula matemática O(1))
  return n * (n + 1) // 2


# Ejemplo de uso:
n = 5
print(
    f"La suma de los primeros {n} números naturales es: {suma_primeros_n(n)}"
)  # Salida: 15 (1 + 2 + 3 + 4 + 5)
```

Desafío 2: Número de elementos pares en una lista
Para contar elementos que cumplen una condición específica, puedes recorrer la lista con un bucle o utilizar una expresión generadora junto con la función sum().

```python
def contar_pares(lista):
  # Opción pitónica usando sum() y una expresión generadora
  return sum(1 for numero in lista if numero % 2 == 0)


# Ejemplo de uso:
mi_lista = [1, 2, 3, 4, 5, 6, 7, 8]
print(
    f"La cantidad de números pares en la lista es: {contar_pares(mi_lista)}"
)  # Salida: 4 (2, 4, 6, 8)
```

Desafío 3: Suma de los primeros $n$ números naturalesEste algoritmo solicita un número entero positivo al usuario y calcula la suma utilizando la fórmula matemática directa, lo cual es muy eficiente ($O(1)$).

```python
def suma_primeros_n():
  try:
    n = int(
        input("Introduce un número entero positivo (n): ")
    )  # Entrada del usuario

    if n <= 0:
      print("El número debe ser un entero positivo mayor a 0.")
      return

    # Fórmula matemática: n * (n + 1) / 2
    resultado = n * (n + 1) // 2
    print(f"La suma de los primeros {n} números naturales es: {resultado}")

  except ValueError:
    print("Por favor, introduce un número entero válido.")


# Ejecutar la función (descomenta la línea de abajo para probarla):
# suma_primeros_n()
```

Desafío 4: Número de elementos pares en una lista
Este algoritmo recibe una serie de números enteros separados por espacios, los convierte en una lista y cuenta cuántos son pares evaluando si su residuo al dividir entre 2 es cero (num % 2 == 0).

```python
def contar_elementos_pares():
  try:
    # Entrada de la lista de números separados por espacios
    entrada = input(
        "Introduce una lista de números enteros separados por espacios (ej: 1 2"
        " 3 4): "
    )
    lista_numeros = [int(x) for x in entrada.split()]

    # Contar los pares usando sum() y una expresión generadora
    pares_totales = sum(1 for num in lista_numeros if num % 2 == 0)

    print(f"Lista ingresada: {lista_numeros}")
    print(f"Cantidad de números pares: {pares_totales}")

  except ValueError:
    print(
        "Error: Asegúrate de ingresar únicamente números enteros separados por"
        " espacios."
    )


# Ejecutar la función (descomenta la línea de abajo para probarla):
# contar_elementos_pares()
```

Desafío 5: Clasificar numeros

Dado tres números distintos, clasificarlos en orden ascendente utilizando comparaciones encadenadas y operadores lógicos.

```python
def clasificar_ascendente(a, b, c):
    """Clasifica tres números distintos en orden ascendente
    utilizando comparaciones encadenadas y operadores lógicos."""

    # Caso 1: a es el menor
    if a < b and b < c:       # Comparación encadenada: a < b < c
        return a, b, c
    elif a < c and c < b:     # a < c < b
        return a, c, b

    # Caso 2: b es el menor
    elif b < a and a < c:     # b < a < c
        return b, a, c
    elif b < c and c < a:     # b < c < a
        return b, c, a

    # Caso 3: c es el menor
    elif c < a and a < b:     # c < a < b
        return c, a, b
    else:                     # c < b < a (por exclusión o comparación directa)
        return c, b, a

# --- Ejemplo de uso ---
n1, n2, n3 = 34, 12, 58
resultado = clasificar_ascendente(n1, n2, n3)

print(f"Números ingresados: {n1}, {n2}, {n3}")
print(f"Orden ascendente: {resultado[0]}, {resultado[1]}, {resultado[2]}")
```

Desafío 6: Calcular promedio
Crear un programa que lea 5 calificaciones y calcule su promedio

```python
def calcular_promedio():
    calificaciones = []
    print("=== Cálculo de Promedio de 5 Calificaciones ===")

    # Bucle para leer las 5 calificaciones
    for i in range(1, 6):
        while True:
            try:
                nota = float(input(f"Ingrese la calificación {i}: "))
                calificaciones.append(nota)
                break
            except ValueError:
                print("❌ Error: Por favor, ingrese un número válido.")

    # Cálculo del promedio
    promedio = sum(calificaciones) / len(calificaciones)

    print(f"\n📊 El promedio final es: {promedio:.2f}")

# Ejecución del programa
if __name__ == "__main__":
    calcular_promedio()
```

Desafío 7: Convertidor de temperaturas
Realizar conversiones de una temperatura dada en Celsius a Fahrenheit y Kelvin utilizando operaciones directas sobre las variables. Los cálculos se realizarán directamente al asignar los valores a las variables.

```python
# Temperatura dada en Celsius
celsius = 25.0

# Conversiones directas al asignar los valores a las nuevas variables
fahrenheit = (celsius * 9/5) + 32
kelvin = celsius + 273.15

# Resultados
print(f"🌡️ Temperatura en Celsius: {celsius}°C")
print(f"🔥 Equivalente en Fahrenheit: {fahrenheit:.2f}°F")
print(f"❄️ Equivalente en Kelvin: {kelvin:.2f}K")
```

Desafío 8: Verificar múltiplos de varios números
Objetivo: Determinar si un número dado es múltiplo de 2, 3, 5, 7, 9, 10, y 11 utilizando comparaciones y operadores lógicos.

*Descripción:*
Crea un programa que, dado un número, verifique si es múltiplo de los números mencionados. Para cada uno de estos números, el programa debe indicar si el número dado es o no un múltiplo de estos.

_Ejemplo de lo que se debe lograr:_
Supongamos que el número dado es 30. El programa deberá determinar:

¿Es 30 múltiplo de 2? Sí

¿Es 30 múltiplo de 3? Sí

¿Es 30 múltiplo de 5? Sí

¿Es 30 múltiplo de 7? No

¿Es 30 múltiplo de 9? No

¿Es 30 múltiplo de 10? Sí

¿Es 30 múltiplo de 11? No

```python
def verificar_multiplos(numero):
    """Determina si un número dado es múltiplo de 2, 3, 5, 7, 9, 10 y 11."""
    # Lista de números divisores a evaluar
    divisores = [2, 3, 5, 7, 9, 10, 11]

    print(f"Analizando el número: {numero}\n")

    for divisor in divisores:
        # Verificamos si el residuo de la división es 0 (múltiplo)
        es_multiplo = (numero % divisor == 0)

        # Convertimos el booleano a "Sí" o "No" para mantener el formato del ejemplo
        respuesta = "Sí" if es_multiplo else "No"

        print(f"¿Es {numero} múltiplo de {divisor}? {respuesta}")

# --- Ejemplo de uso con el número 30 ---
verificar_multiplos(30)
```

Desafío 10: Calculadora Simple
Escriba un programa que pida al usuario que ingrese dos números y luego imprima la suma, la resta, la multiplicación y la división de esos números.

```python
def calculadora_simple():
    print("=== Calculadora Simple ===")

    try:
        # Solicitar los dos números al usuario
        num1 = float(input("Ingrese el primer número: "))
        num2 = float(input("Ingrese el segundo número: "))

        # Realizar las operaciones básicas
        suma = num1 + num2
        resta = num1 - num2
        multiplicacion = num1 * num2

        # Mostrar los resultados
        print(f"\nResultados:")
        print(f"  - Suma: {num1} + {num2} = {suma}")
        print(f"  - Resta: {num1} - {num2} = {resta}")
        print(f"  - Multiplicación: {num1} * {num2} = {multiplicacion}")

        # Controlar la división por cero
        if num2 != 0:
            division = num1 / num2
            print(f"  - División: {num1} / {num2} = {division:.2f}")
        else:
            print("  - División: ⚠️ No es posible dividir entre cero.")

    except ValueError:
        print("❌ Error: Por favor, ingrese únicamente valores numéricos válidos.")

# Ejecución del programa
if __name__ == "__main__":
    calculadora_simple()
```

Desafío 11: Formateo de Factura
Escriba un programa que pueda tomar los detalles de los productos (nombre, cantidad, precio) y produzca una factura bien formateada.

```python
def generar_factura():
    print("=== Generador de Facturas ===")
    items = []

    # Bucle para registrar productos de forma interactiva
    while True:
        nombre = input("Ingrese el nombre del producto (o presione Enter para finalizar): ").strip()
        if not nombre:
            break

        try:
            cantidad = int(input(f"Ingrese la cantidad de '{nombre}': "))
            precio = float(input(f"Ingrese el precio unitario de '{nombre}': "))

            if cantidad <= 0 or precio < 0:
                print("⚠️ La cantidad debe ser mayor a 0 y el precio no puede ser negativo.")
                continue

            items.append({"nombre": nombre, "cantidad": cantidad, "precio": precio})
            print(f"✅ '{nombre}' agregado correctamente.\n")

        except ValueError:
            print("❌ Error: Ingrese valores numéricos válidos para la cantidad y el precio.\n")

    if not items:
        print("\nℹ️ No se registraron productos para generar la factura.")
        return

    # --- Impresión de la Factura Formateada ---
    print("\n" + "=" * 58)
    print(f"{'FACTURA DE COMPRA':^58}")
    print("=" * 58)
    # Encabezados con alineación fija
    print(f"{'Producto':<22} | {'Cant':<6} | {'P. Unit':<10} | {'Subtotal':<10}")
    print("-" * 58)

    total_general = 0
    for item in items:
        subtotal = item["cantidad"] * item["precio"]
        total_general += subtotal
        # Formateo con aliniación y control de decimales
        print(f"{item['nombre']:<22} | {item['cantidad']:<6} | ${item['precio']:<9.2f} | ${subtotal:<9.2f}")

    print("-" * 58)
    print(f"{'TOTAL A PAGAR:':<43} ${total_general:>12.2f}")
    print("=" * 58)

# Ejecución del programa
if __name__ == "__main__":
    generar_factura()
```
