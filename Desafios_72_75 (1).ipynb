Desafío extra: Determinar el mayor de tres números enteros.

```python
def determinar_mayor():
    print("--- Desafío Extra: Mayor de tres números ---")
    try:
        # Solicitamos los tres números enteros
        num1 = int(input("Ingresa el primer número entero: "))
        num2 = int(input("Ingresa el segundo número entero: "))
        num3 = int(input("Ingresa el tercer número entero: "))

        # 1. Comprobamos si los tres números son iguales
        if num1 == num2 and num2 == num3:
            print(f"Los tres números son iguales ({num1}).")

        # 2. Comparamos para encontrar el mayor (o si hay empates entre dos)
        elif num1 >= num2 and num1 >= num3:
            if num1 == num2:
                print(f"El número mayor es {num1} (empate entre el primero y el segundo).")
            elif num1 == num3:
                print(f"El número mayor es {num1} (empate entre el primero y el tercero).")
            else:
                print(f"El número mayor es: {num1}")

        elif num2 >= num1 and num2 >= num3:
            if num2 == num3:
                print(f"El número mayor es {num2} (empate entre el segundo y el tercero).")
            else:
                print(f"El número mayor es: {num2}")

        else:
            print(f"El número mayor es: {num3}")

    except ValueError:
        print("❌ Error: Debes ingresar únicamente números enteros válidos.")

# Ejecutamos la función
determinar_mayor()
```

Utilizamos estructuras de control condicionales (if, elif, else) y operadores lógicos (and) para evaluar todas las combinaciones posibles de mayor tamaño e igualdad entre los números.

Desafío 72: Buscar un código de producto y su posición.

```python
def buscar_codigo_producto():
    print("\n--- Desafío 72: Sistema de Inventario ---")

    # Lista de códigos de productos disponibles en el inventario
    inventario = ["PROD-001", "PROD-045", "PROD-102", "PROD-205", "PROD-310"]

    print(f"Códigos actuales en el sistema: {inventario}")

    # Solicitamos el código al usuario (lo convertimos a mayúsculas para evitar problemas de tipeo)
    codigo_buscado = input("Ingresa el código del producto a buscar (ej. PROD-102): ").strip().upper()

    # Verificamos si el código está en la lista
    if codigo_buscado in inventario:
        # Obtenemos la posición exacta (índice) donde aparece por primera vez
        posicion = inventario.index(codigo_buscado)
        print(f"✅ ¡Código encontrado! El producto '{codigo_buscado}' está en la **posición {posicion}** (índice base 0) de la lista.")
        print(f"   (En orden humano, ocupa el puesto #{posicion + 1}).")
    else:
        print(f"❌ El código '{codigo_buscado}' no se ha encontrado en el inventario.")

# Ejecutamos la función
buscar_codigo_producto()
```

El operador in valida de forma rápida y limpia si el elemento pertenece a la lista, y el método .index() nos devuelve el número de índice exacto de la primera coincidencia sin necesidad de recorrer la lista manualmente con un bucle for.


Desafío 73:
Tienes una lista de números en la que algunos elementos están repetidos. Desarrolla un programa que elimine todos los elementos duplicados y deje únicamente una aparición de cada uno. La salida debe mostrar la lista original y la lista sin duplicados.

```python
# Definimos una lista con números repetidos
lista_original = [4, 2, 9, 4, 1, 2, 8, 9, 3, 1]

# Eliminamos duplicados manteniendo el orden original
# dict.fromkeys() elimina las claves duplicadas y list() vuelve a convertirlo en lista
lista_sin_duplicados = list(dict.fromkeys(lista_original))

# Mostramos los resultados en pantalla
print(f"Lista original:     {lista_original}")
print(f"Lista sin duplicados: {lista_sin_duplicados}")
```

Asi funciona:
dict.fromkeys(lista_original): Crea un diccionario temporal donde los elementos de la lista se usan como claves. Como los diccionarios no permiten claves duplicadas, los elementos repetidos se eliminan automáticamente.

list(...): Vuelve a transformar ese diccionario resultante en una lista tradicional para que podamos trabajar con ella como una lista común.

Alternativa con set: Si el orden de los elementos no te importa, también podrías usar list(set(lista_original)), aunque el orden final podría cambiar al azar debido a cómo los conjuntos almacenan los datos internamente.

Desafío 74: Sumar listas de diferentes longitudes.

```python
from itertools import zip_longest

# Definimos dos listas de diferentes longitudes
lista1 = [1, 2, 3, 4, 5]
lista2 = [10, 20, 30]

# Usamos zip_longest con fillvalue=0 para considerar 0 si falta un elemento
suma_listas = [x + y for x, y in zip_longest(lista1, lista2, fillvalue=0)]

print(f"Lista 1: {lista1}")
print(f"Lista 2: {lista2}")
print(f"Suma por posiciones: {suma_listas}")
```

Funciona de la siguiente forma:
zip_longest(..., fillvalue=0): Toma los elementos de ambas listas en paralelo (posición por posición). Si una lista se acaba antes, en lugar de detenerse o dar error, asigna un 0 al valor faltante.

List Comprehension ([x + y for ...]): Recorre esos pares generados y suma los valores de manera limpia y eficiente en una sola línea.

Desafío 75: Potencias de 2 con List Comprehension.

```python
# Creamos la lista utilizando una list comprehension y el operador de potencia (**)
potencias = [2**i for i in range(10)]

print("Lista de potencias de 2 (del 0 al 9):")
print(potencias)
```

Funciona de la siguiente forma:
range(10): Genera una secuencia de números enteros desde el 0 hasta el 9.2**i: Eleva el número 2 a la potencia de cada valor generado por el range.El resultado directo es una lista que va desde $2^0$ (que es 1) hasta $2^9$ (que es 512).
