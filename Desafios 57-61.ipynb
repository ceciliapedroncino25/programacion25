Desafío 57: Herencia — clase Poeta
En el sistema de biblioteca ya contamos con una clase Autor que almacena información básica como nombre y nacionalidad.
Ahora queremos representar un tipo particular de autor: un poeta.
Implementa una clase Poeta que:

Herede de la clase Autor.
Agregue un nuevo atributo llamado tipo_poesia.
Utilice super() para inicializar los atributos heredados.
Incluya un método que muestre toda la información del poeta: nombre, nacionalidad y tipo de poesía.
Luego:

Crea al menos dos objetos Poeta.
Muestra sus datos.
Verifica que los objetos puedan utilizar también los atributos y métodos heredados de Autor.
Pregunta final: ¿qué ventaja tiene crear Poeta como subclase de Autor en lugar de volver a definir desde cero los atributos y métodos comunes?

```python
# Clase Padre
class Autor:
    def __init__(self, nombre, nacionalidad):
        self.nombre = nombre
        self.nacionalidad = nacionalidad

    def obtener_informacion_basica(self):
        return f"Autor: {self.nombre} | Nacionalidad: {self.nacionalidad}"


# Clase Hija (Subclase)
class Poeta(Autor):
    def __init__(self, nombre, nacionalidad, tipo_poesia):
        # Inicializa los atributos de la clase padre
        super().__init__(nombre, nacionalidad)
        # Atributo propio de la subclase
        self.tipo_poesia = tipo_poesia

    def mostrar_informacion(self):
        return f"Poeta: {self.nombre} | Nacionalidad: {self.nacionalidad} | Tipo de Poesía: {self.tipo_poesia}"


# 1. Creación de dos objetos Poeta
poeta1 = Poeta("Pablo Neruda", "Chilena", "Lírica")
poeta2 = Poeta("Dante Alighieri", "Italiana", "Épica")

# 2. Muestra de sus datos
print("--- Datos completos del poeta ---")
print(poeta1.mostrar_informacion())
print(poeta2.mostrar_informacion())

# 3. Verificación de herencia (uso de atributos y métodos de Autor)
print("\n--- Verificación de métodos y atributos heredados ---")
print(f"Atributo heredado directamente: {poeta1.nombre}")
print(f"Método heredado de Autor: {poeta2.obtener_informacion_basica()}")
```


Pregunta final: ¿qué ventaja tiene crear Poeta como subclase de Autor en lugar de volver a definir desde cero los atributos y métodos comunes?

Reutilización de código (DRY - Don't Repeat Yourself): Evitas duplicar la lógica de asignación para nombre y nacionalidad. Si la clase Autor cambia en el futuro (por ejemplo, si agregas un atributo fecha_nacimiento), Poeta lo heredará automáticamente sin modificar su propio código.

Mantenibilidad y consistencia: Garantiza que todos los tipos de autores (poetas, novelistas, ensayistas) compartan una estructura base uniforme dentro del sistema de la biblioteca.

Polimorfismo: Permite tratar a las instancias de Poeta genéricamente como objetos Autor en funciones que gestionen libros, catálogos o búsquedas en la biblioteca.

Desafío 58:
Crea una clase Bibliotecario que herede de Usuario y tenga atributos específicos como sección y años_experiencia.

```python
# Clase Padre
class Usuario:
    def __init__(self, nombre, id_usuario):
        self.nombre = nombre
        self.id_usuario = id_usuario

    def obtener_identificacion(self):
        return f"Usuario: {self.nombre} (ID: {self.id_usuario})"


# Clase Hija (Subclase)
class Bibliotecario(Usuario):
    def __init__(self, nombre, id_usuario, seccion, anos_experiencia):
        # Inicialización de atributos heredados de Usuario
        super().__init__(nombre, id_usuario)
        # Atributos específicos de Bibliotecario
        self.seccion = seccion
        self.anos_experiencia = anos_experiencia

    def mostrar_detalle_personal(self):
        return (
            f"Bibliotecario: {self.nombre} | ID: {self.id_usuario} | "
            f"Sección: {self.seccion} | Experiencia: {self.anos_experiencia} años"
        )


# Ejemplo de uso y prueba
bibliotecario1 = Bibliotecario("Laura Gómez", "B-104", "Colecciones Especiales", 8)

# 1. Mostrar detalles con el método propio
print(bibliotecario1.mostrar_detalle_personal())

# 2. Demostración de herencia (acceso a atributo y método de la clase padre)
print(f"Atributo heredado (Nombre): {bibliotecario1.nombre}")
print(f"Método heredado: {bibliotecario1.obtener_identificacion()}")
```

Es clave:

super().__init__(nombre, id_usuario): Pasa la inicialización del nombre y la identificación a la clase base Usuario.

Atributos de subclase: seccion y anos_experiencia extienden la funcionalidad sin modificar la clase base.

Desafío 59:
Diseña una clase LibroDigital que herede de Libro y añada atributos como formato (e.g., PDF, EPUB) y tamaño_archivo. Además, implementa una subclase EBook que sobrescriba un método para mostrar información específica, como enlaces de descarga.

```python
# Clase Base (Padre)
class Libro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor

    def mostrar_informacion(self):
        return f"Libro: '{self.titulo}' - Autor: {self.autor}"


# Clase Intermedia (Hereda de Libro)
class LibroDigital(Libro):
    def __init__(self, titulo, autor, formato, tamano_archivo):
        # Inicializa los atributos de la clase Libro
        super().__init__(titulo, autor)
        # Atributos específicos de LibroDigital
        self.formato = formato.upper()          # e.g., PDF, EPUB
        self.tamano_archivo = tamano_archivo   # e.g., en MB

    def mostrar_informacion(self):
        # Combina la información base con los detalles digitales
        info_base = super().mostrar_informacion()
        return f"{info_base} | Formato: {self.formato} | Tamaño: {self.tamano_archivo} MB"


# Subclase Específica (Hereda de LibroDigital)
class EBook(LibroDigital):
    def __init__(self, titulo, autor, formato, tamano_archivo, enlace_descarga):
        # Inicializa los atributos heredados de LibroDigital y Libro
        super().__init__(titulo, autor, formato, tamano_archivo)
        # Atributo propio de EBook
        self.enlace_descarga = enlace_descarga

    def mostrar_informacion(self):
        # Sobrescribe el método para añadir el enlace de descarga específico
        info_digital = super().mostrar_informacion()
        return f"[eBook] {info_digital} | Descarga: {self.enlace_descarga}"


# --- Pruebas de funcionamiento ---

# 1. Instancia de LibroDigital
libro_digital = LibroDigital("Don Quijote de la Mancha", "Miguel de Cervantes", "PDF", 4.5)
print("--- Libro Digital ---")
print(libro_digital.mostrar_informacion())

# 2. Instancia de EBook
ebook = EBook(
    titulo="Cien Años de Soledad",
    autor="Gabriel García Márquez",
    formato="EPUB",
    tamano_archivo=2.1,
    enlace_descarga="https://biblioteca.org/download/cien-anos-soledad.epub"
)

print("\n--- EBook (con sobrescritura de método) ---")
print(ebook.mostrar_informacion())

# 3. Verificación de acceso a atributos de todas las clases ancestro
print("\n--- Acceso directo a atributos del EBook ---")
print(f"Título (Libro): {ebook.titulo}")
print(f"Formato (LibroDigital): {ebook.formato}")
print(f"Enlace (EBook): {ebook.enlace_descarga}")
```

Se aplica:

Herencia multinivel: EBook hereda de LibroDigital, que a su vez hereda de Libro. EBook tiene acceso a todos los atributos y métodos de la cadena.

Sobrescritura de métodos (Overriding): Tanto LibroDigital como EBook redefinen el método mostrar_informacion().

Uso de super(): Permite reutilizar la lógica de formateo del método en la clase superior y simplemente concatenar los nuevos datos de la subclase.

Desafío 60:
Implementa una clase EscritorAcademico que herede de Escritor y Academico, e incluya un método adicional para publicar artículos académicos. Asegúrate de utilizar correctamente la función super() para inicializar las clases base.


```python
# Primera clase base
class Escritor:
    def __init__(self, estilo, **kwargs):
        # super().__init__ pasa los argumentos sobrantes a la siguiente clase en el MRO
        super().__init__(**kwargs)
        self.estilo = estilo

    def escribir_obra(self, titulo):
        return f"Escribiendo la obra '{titulo}' en estilo {self.estilo}."


# Segunda clase base
class Academico:
    def __init__(self, universidad, campo_estudio, **kwargs):
        super().__init__(**kwargs)
        self.universidad = universidad
        self.campo_estudio = campo_estudio

    def investigar(self, tema):
        return f"Investigando el tema '{tema}' en el campo de {self.campo_estudio} ({self.universidad})."


# Clase con herencia múltiple
class EscritorAcademico(Escritor, Academico):
    def __init__(self, estilo, universidad, campo_estudio, universidad_publicacion):
        # super() coordina la inicialización de Escritor y Academico
        super().__init__(
            estilo=estilo,
            universidad=universidad,
            campo_estudio=campo_estudio
        )
        self.universidad_publicacion = universidad_publicacion

    # Método específico solicitado
    def publicar_articulo_academico(self, titulo_articulo, revista):
        return (
            f"El artículo '{titulo_articulo}' fue publicado en la revista '{revista}' "
            f"con el respaldo de la {self.universidad_publicacion}."
        )


# --- Pruebas de funcionamiento ---

# Instancia de EscritorAcademico
investigador = EscritorAcademico(
    estilo="Ensayo argumentativo",
    universidad="Universidad de la República",
    campo_estudio="Ciencias de la Computación",
    universidad_publicacion="UdelaR Press"
)

# 1. Uso de métodos heredados de Escritor
print(investigador.escribir_obra("Impacto de la IA en el Código Moderno"))

# 2. Uso de métodos heredados de Academico
print(investigador.investigar("Algoritmos de Herencia Múltiple"))

# 3. Uso del método propio de EscritorAcademico
print(investigador.publicar_articulo_academico(
    "MRO y super() en Python",
    "Journal of Software Architecture"
))

# 4. Verificación del orden de resolución de métodos (MRO)
print("\nOrden de resolución de métodos (MRO):")
for clase in EscritorAcademico.__mro__:
    print(f"- {clase.__name__}")
```

Hay que tener en cuenta sobre Herencia Múltiple y super()MRO (Method Resolution Order): Python resuelve qué métodos ejecutar siguiendo una jerarquía ordenada. Para EscritorAcademico, el orden es: EscritorAcademico $\rightarrow$ Escritor $\rightarrow$ Academico $\rightarrow$ object.Argumentos nombrados (**kwargs): Para que super().__init__() funcione correctamente a lo largo de toda la cadena sin fallar en las clases intermedias, es una buena práctica usar **kwargs. De esta forma, cada clase consume los parámetros que necesita y delega el resto a la siguiente clase base en el MRO.

Desafío 61:
Crea una jerarquía de clases para representar diferentes tipos de empleados en una biblioteca, utilizando herencia múltiple y composición. Por ejemplo, implementa clases como Empleado, Gerente, Tecnico, y Voluntario, donde Gerente y Tecnico hereden de Empleado, y algunos puedan tener roles adicionales mediante composición con otras clases como Administrador o Mantenimiento.

```python
# --------------------------------------------------
# 1. Clases para Composición (Roles Adicionales)
# --------------------------------------------------
class RolAdministrador:
    def gestionar_presupuesto(self, monto: float):
        print(f"  [Administración] Gestionando presupuesto de ${monto:,.2f}")

    def aprobar_compras(self, item: str):
        print(f"  [Administración] Compra aprobada para: {item}")


class RolMantenimiento:
    def reparar_equipo(self, equipo: str):
        print(f"  [Mantenimiento] Reparando el equipo: {equipo}")

    def inspeccionar_instalaciones(self):
        print("  [Mantenimiento] Inspeccionando salas y estanterías")


# --------------------------------------------------
# 2. Clases Base y Derivadas (Herencia Simple)
# --------------------------------------------------
class Empleado:
    def __init__(self, nombre: str, id_empleado: str, salario: float):
        self.nombre = nombre
        self.id_empleado = id_empleado
        self.salario = salario
        # Roles asignados por composición
        self.roles_adicionales = []

    def agregar_rol(self, rol):
        """Asigna una responsabilidad adicional mediante composición."""
        self.roles_adicionales.append(rol)
        print(f"Rol {rol.__class__.__name__} asignado a {self.nombre}")

    def mostrar_info(self):
        print(f"[{self.id_empleado}] {self.nombre} - Salario: ${self.salario:,.2f}")


class Gerente(Empleado):
    def __init__(self, nombre: str, id_empleado: str, salario: float, departamento: str):
        super().__init__(nombre, id_empleado, salario)
        self.departamento = departamento

    def supervisar_equipo(self):
        print(f"{self.nombre} está supervisando el departamento de {self.departamento}")


class Tecnico(Empleado):
    def __init__(self, nombre: str, id_empleado: str, salario: float, especialidad: str):
        super().__init__(nombre, id_empleado, salario)
        self.especialidad = especialidad

    def realizar_soporte(self):
        print(f"{self.nombre} está realizando soporte en {self.especialidad}")


# --------------------------------------------------
# 3. Herencia Múltiple
# --------------------------------------------------
class GerenteTecnico(Gerente, Tecnico):
    """
    Combina responsabilidades de Gerente y Técnico mediante herencia múltiple.
    """
    def __init__(self, nombre: str, id_empleado: str, salario: float, departamento: str, especialidad: str):
        # Inicialización directa para evitar ambigüedades en la cadena MRO
        Empleado.__init__(self, nombre, id_empleado, salario)
        self.departamento = departamento
        self.especialidad = especialidad


# --------------------------------------------------
# 4. Clase Independiente (Voluntario)
# --------------------------------------------------
class Voluntario:
    def __init__(self, nombre: str, horas_disponibles: int):
        self.nombre = nombre
        self.horas_disponibles = horas_disponibles
        self.roles_adicionales = []

    def agregar_rol(self, rol):
        self.roles_adicionales.append(rol)

    def organizar_eventos(self):
        print(f"El voluntario {self.nombre} está organizando un club de lectura.")


# --------------------------------------------------
# Ejemplo de uso
# --------------------------------------------------
if __name__ == "__main__":
    # 1. Uso de Herencia Simple + Composición
    gerente_biblio = Gerente("Laura Gómez", "EMP-001", 3500.0, "Colecciones Especiales")
    gerente_biblio.mostrar_info()
    gerente_biblio.supervisar_equipo()

    # Composición: Se asigna el rol de Administrador
    gerente_biblio.agregar_rol(RolAdministrador())
    gerente_biblio.roles_adicionales[0].gestionar_presupuesto(15000.0)
    print("-" * 50)

    # 2. Uso de Herencia Múltiple + Composición
    lider_it = GerenteTecnico("Carlos Ruiz", "EMP-002", 4200.0, "Sistemas", "Redes y Servidores")
    lider_it.mostrar_info()
    lider_it.supervisar_equipo()    # Método heredado de Gerente
    lider_it.realizar_soporte()     # Método heredado de Técnico

    # Composición: Se asigna el rol de Mantenimiento
    lider_it.agregar_rol(RolMantenimiento())
    lider_it.roles_adicionales[0].reparar_equipo("Servidor del Catálogo OPAC")
    print("-" * 50)

    # 3. Voluntario con Composición
    voluntario = Voluntario("Ana Sola", 10)
    voluntario.organizar_eventos()
    voluntario.agregar_rol(RolMantenimiento())
    voluntario.roles_adicionales[0].inspeccionar_instalaciones()
```

Para modelar una jerarquía de empleados de biblioteca utilizando herencia múltiple y composición en Python, implementamos:

Clases de Composición (Roles Adicionales): Representan habilidades o responsabilidades secundarias que se asignan a un empleado.

Jerarquía Base y Derivadas (Herencia): La clase base Empleado, de la cual heredan Gerente y Tecnico. Voluntario no hereda de Empleado (ya que no es un empleado remunerado), pero puede compartir interfaces.

Herencia Múltiple: Una clase como GerenteTecnico que combina habilidades de gestión y técnicas.
