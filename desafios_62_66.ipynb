Desafío 62:
Crea una clase Musico que tenga un método instrumento y crea dos subclases Guitarrista y Baterista que sobrescriban el método instrumento. Instancia objetos de estas clases y demuestra el polimorfismo.

```python
from abc import ABC, abstractmethod

class Musico(ABC):
    @abstractmethod
    def instrumento(self):
        pass

class Guitarrista(Musico):
    def instrumento(self):
        return "Toco la guitarra eléctrica."

class Baterista(Musico):
    def instrumento(self):
        return "Toco la batería."

# Demostración
orquesta = [Guitarrista(), Baterista()]
for m in orquesta:
    print(m.instrumento())
```

¿Cómo funciona el polimorfismo en esta situación?
Sobrescritura de métodos: Tanto Guitarrista como Baterista heredan de la clase base Musico, pero cada una adapta el método instrumento() a su propia realidad.

Comportamiento dinámico: La función presentar_musico acepta cualquier objeto que sea un Musico (o sus subclases). Al llamarlo, Python determina automáticamente qué versión del método instrumento() ejecutar según el tipo de objeto real en tiempo de ejecución, sin importar que la función espere un tipo genérico.

Desafío 63:
Añade un método biografia a la clase Autor y sobrescríbelo en la clase Escritor. Instancia un objeto de la clase Escritor y muestra cómo se puede acceder al método biografia de ambas clases.

```python
class Autor:
    def biografia(self):
        return "Autor: Persona que crea obras escritas y de investigación."

class Escritor(Autor):
    def biografia(self):
        # 1. Accedemos al método del padre usando super()
        biografia_padre = super().biografia()
        return f"{biografia_padre} Específicamente, soy un escritor de novelas de ficción y misterio."

# Demostración
if __name__ == "__main__":
    # Instanciamos un objeto de la clase Escritor
    mi_escritor = Escritor()

    print("--- 1. Accediendo al método sobrescrito (que incluye al padre con super()) ---")
    print(mi_escritor.biografia())

    print("\n--- 2. Accediendo explícitamente al método original de la clase Autor ---")
    # Llamando al método de la clase base pasando la instancia del escritor
    print(Autor.biografia(mi_escritor))
```

¿Cómo se accede a ambos métodos?
Mediante super(): Dentro de la subclase Escritor, la función super().biografia() te permite invocar y reutilizar el código del método original de la clase Autor.

De forma explícita (Autor.biografia(instancia)): Python permite invocar el método de la clase padre directamente pasando el objeto como argumento (self), lo que resulta útil si necesitas consultar la versión original sin perder el método sobrescrito.

Desafío 64:
En este desafío, vamos a extender la clase Libro para crear una subclase `LibroEspecializado`. Un `LibroEspecializado`, además de tener un título y un autor, también tiene un campo de estudio y un nivel de especialización (básico, intermedio, avanzado).

```python
class Libro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor

    def descripcion(self):
        return f"Libro: '{self.titulo}' escrito por {self.autor}"

class LibroEspecializado(Libro):
    def __init__(self, titulo, autor, campo_estudio, nivel_especializacion):
        # Heredamos los atributos de la clase padre usando super()
        super().__init__(titulo, autor)
        self.campo_estudio = campo_estudio
        self.nivel_especializacion = nivel_especializacion

    # Sobrescribimos o extendemos el método descripcion para incluir los nuevos campos
    def descripcion(self):
        descripcion_base = super().descripcion()
        return (f"{descripcion_base}\n"
                f"Campo de estudio: {self.campo_estudio}\n"
                f"Nivel: {self.nivel_especializacion.capitalize()}")

# Demostración
if __name__ == "__main__":
    # Instanciamos un libro especializado
    mi_libro = LibroEspecializado(
        titulo="Inteligencia Artificial: Un Enfoque Moderno",
        autor="Stuart Russell y Peter Norvig",
        campo_estudio="Ciencias de la Computación",
        nivel_especializacion="avanzado"
    )

    print("--- Detalles del Libro Especializado ---")
    print(mi_libro.descripcion())
```

Clave de esta implementación:
Uso de super(): Permite reutilizar el constructor (__init__) y los métodos de la clase padre (Libro) para evitar repetir código.

Extensión de atributos: La subclase añade atributos específicos (campo_estudio y nivel_especializacion) adaptándose a necesidades más complejas sin romper la estructura original.

Polimorfismo: El método descripcion() se sobrescribe para enriquecer la información que se muestra al usuario.

Desafío 65: Polimorfismo con figuras geométricas
Queremos modelar distintas figuras geométricas que comparten una misma operación: calcular su área.
Crea una clase base Figura que defina un método area().
Luego implementa dos subclases:

Circulo, con un atributo radio.
Cuadrado, con un atributo lado.
Cada subclase deberá:

Heredar de Figura.
Sobrescribir el método area() para calcular el área correspondiente.
Utilizar los atributos propios de cada figura.
Finalmente:

Crea al menos un objeto Circulo y un objeto Cuadrado.
Guarda ambos objetos en una misma lista.
Recorre la lista y ejecuta area() sobre cada objeto sin preguntar de qué tipo de figura se trata.
Muestra el nombre de la figura y el área calculada.
Pregunta final: ¿por qué podemos llamar al mismo método area() sobre objetos distintos y obtener comportamientos diferentes?

```python
import math

class Figura:
    def area(self):
        """Método base que será sobrescrito por las subclases"""
        pass

class Circulo(Figura):
    def __init__(self, radio):
        self.radio = radio

    def area(self):
        return math.pi * (self.radio ** 2)

class Cuadrado(Figura):
    def __init__(self, lado):
        self.lado = lado

    def area(self):
        return self.lado ** 2

# Demostración
if __name__ == "__main__":
    # 1. Creamos objetos de ambas clases
    mi_circulo = Circulo(radio=5)
    mi_cuadrado = Cuadrado(lado=4)

    # 2. Los guardamos en una misma lista
    figuras = [mi_circulo, mi_cuadrado]

    print("--- Demostración de Polimorfismo con Figuras ---")
    # 3. Recorremos la lista y ejecutamos area() sin importar el tipo
    for figura in figuras:
        nombre_figura = type(figura).__name__
        resultado_area = figura.area()
        print(f"Figura: {nombre_figura} -> Área: {resultado_area:.2f}")
```

¿Por qué podemos llamar al mismo método area() sobre objetos distintos y obtener comportamientos diferentes?

Esto es posible gracias al polimorfismo y al enlace dinámico (dynamic dispatch) en tiempo de ejecución:

Sobrescritura de métodos (Overriding): Tanto Circulo como Cuadrado heredan de la clase Figura, pero cada una implementa su propia fórmula matemática dentro del método area().

Independencia del tipo: Al recorrer la lista, Python no necesita evaluar con un condicional (if/else) de qué clase es cada objeto. Simplemente le ordena al objeto que ejecute su método area().

Resolución en tiempo de ejecución: Python identifica automáticamente el tipo real del objeto (Circulo o Cuadrado) en el momento exacto en que se ejecuta la línea y llama a la versión correspondiente del método, logrando que un mismo mensaje (area()) produzca respuestas completamente distintas según quién lo reciba.

 Desafío 66: Polimorfismo en operaciones matemáticas
En este desafío, aplicarás el polimorfismo para realizar diferentes operaciones matemáticas. Deberás crear una clase base Operacion con un método resultado y dos subclases Suma y Multiplicacion que sobrescriban este método para realizar las operaciones correspondientes.

```python
class Operacion:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def resultado(self):
        """Método base que será sobrescrito por las subclases"""
        pass

class Suma(Operacion):
    def resultado(self):
        return self.a + self.b

class Multiplicacion(Operacion):
    def resultado(self):
        return self.a * self.b

# Demostración de Polimorfismo
if __name__ == "__main__":
    # Creamos objetos de distintas operaciones pasando los operandos
    operaciones = [
        Suma(10, 5),
        Multiplicacion(4, 3),
        Suma(2.5, 7.5)
    ]

    print("--- Demostración de Polimorfismo Matemático ---")
    # Recorremos la lista y ejecutamos .resultado() de forma uniforme
    for op in operaciones:
        tipo_operacion = type(op).__name__
        res = op.resultado()
        print(f"Operación: {tipo_operacion} (con {op.a} y {op.b}) -> Resultado: {res}")
```

¿Cómo funciona el polimorfismo en este desafío?
Estructura común: Tanto Suma como Multiplicacion heredan de Operacion, compartiendo la misma estructura para recibir los dos números (a y b) a través del constructor.

Comportamiento específico: Cada subclase implementa su propia lógica dentro del método resultado().

Ejecución uniforme: Al igual que en las figuras geométricas, la lista agrupa diferentes objetos matemáticos y llama al método resultado() sin preocuparse por cuál operación matemática se está ejecutando en cada iteración.
